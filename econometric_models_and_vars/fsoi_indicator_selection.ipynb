```python
# Data from https://www.harita.gov.tr/il-ve-ilce-yuzolcumleri

data_HGM = pd.read_excel("Türkiye_Municipal_Areas.xlsx")
data_HGM.head(10)
```

```python
# Deleting town area values, keeping only city area values

data_HGM = data_HGM[
    data_HGM["İL VE İLÇE YÜZ ÖLÇÜMLERİ"].isna() &
    data_HGM["Unnamed: 1"].isna() &
    data_HGM["Unnamed: 2"].notna()]

data_HGM.reset_index(inplace=True, drop=True)
print(data_HGM)
```

```python
import unicodedata

# Correction of city name values for Türkiye-HGM data

data_HGM = data_HGM.rename(columns={"Unnamed: 2" : "Location_Name", 
                        "Unnamed: 3" : "Area_km2"})

data_HGM = data_HGM.drop(columns=["İL VE İLÇE YÜZ ÖLÇÜMLERİ", "Unnamed: 1"])
data_HGM.drop(index=0, axis=0, inplace=True)
data_HGM.reset_index(inplace=True, drop=True)


# Capitalize city names while preserving Turkish encoding
def normalize_turkish(text):
    if not isinstance(text, str):
        return text
    # Normalize Unicode composition (fixes dotted İ etc.)
    text = unicodedata.normalize("NFKC", text.strip())
    # Replace locale-specific characters before capitalization
    text = text.translate(str.maketrans("IİŞĞÜÇÖ", "ıişğüçö")).capitalize()
    return text

# Apply to city names
data_HGM["Location_Name"] = data_HGM["Location_Name"].apply(normalize_turkish)


print(data_HGM["Location_Name"].unique())
# For Hakkari, İstanbul, and İzmir
data_HGM["Location_Name"] = data_HGM["Location_Name"].str.replace("Hakkâri", "Hakkari")
data_HGM["Location_Name"] = data_HGM["Location_Name"].str.replace("Istanbul", "İstanbul")
data_HGM["Location_Name"] = data_HGM["Location_Name"].str.replace("Izmir", "İzmir")
print(data_HGM["Location_Name"].unique())

```

```python
data_official_Türkiye = data_TÜİK.merge(
    data_HGM.assign(Location_lower=data_HGM["Location_Name"].str.lower())[["Location_lower", "Area_km2"]],
    left_on=data_TÜİK["Location_Name"].str.lower(),
    right_on="Location_lower",
    how="left"
).drop(columns="Location_lower")
data_official_Türkiye["Area_km2"] = data_official_Türkiye["Area_km2"].astype("float64")
```

```python
# Control empty area values

missing_area = data_official_Türkiye[data_official_Türkiye["Area_km2"].isna()]
print(missing_area["Location_Name"].unique())

# Because that I replaced the city encodings, only Türkiye was left, picked a reference year 2008 

total_area = data_official_Türkiye.loc[
    (data_official_Türkiye["Year"] == 2008) & (data_official_Türkiye["Location_Name"] != "Türkiye"), "Area_km2"].sum()

# Assign total to Türkiye
data_official_Türkiye.loc[data_official_Türkiye["Location_Name"] == "Türkiye", "Area_km2"] = total_area

missing_area = data_official_Türkiye[data_official_Türkiye["Area_km2"].isna()]
print(missing_area["Location_Name"].unique())


```

```python
law6360_cities = ["Aydın", "Balıkesir", "Denizli", "Hatay", "Malatya", "Manisa", "Mardin", "Muğla", "Kahramanmaraş", "Ordu", "Tekirdağ", "Trabzon", "Şanlıurfa", "Van"]
all_metropolitan_cities = ["Adana", "Ankara", "Antalya", "Aydın", "Balıkesir", "Bursa", "Denizli", "Diyarbakır",
                           "Erzurum", "Eskişehir", "Gaziantep", "Hatay", "İstanbul", "İzmir", "Kahramanmaraş",
                           "Kayseri", "Kocaeli", "Konya", "Malatya", "Manisa", "Mardin", "Mersin", "Muğla", "Ordu",
                           "Sakarya", "Samsun", "Şanlıurfa", "Tekirdağ", "Trabzon"]
old_metropol_cities = [c for c in all_metropolitan_cities if c not in law6360_cities]

data_official_Türkiye["Treated"] = 0
data_official_Türkiye.loc[data_official_Türkiye["Location_Name"].isin(law6360_cities), "Treated"] = 1
data_official_Türkiye.loc[data_official_Türkiye["Location_Name"].isin(old_metropol_cities), "Treated"] = 2
data_official_Türkiye.loc[data_official_Türkiye["Location_Name"] == "Türkiye", "Treated"] = 3  # national aggregate row, not a real city

data_official_Türkiye["Treated"] = pd.Categorical(data_official_Türkiye["Treated"], categories=[0, 1, 2, 3], ordered=True)

treated_labels = {0: "Non-metropolitan", 1: "New-metropolitan (2012)", 2: "Old-metropolitan", 3: "Türkiye"}
data_official_Türkiye["Treated_Label"] = pd.Categorical(
    data_official_Türkiye["Treated"].map(treated_labels),
    categories=list(treated_labels.values()),
    ordered=True
)
data_official_Türkiye.columns
```

```python
# Transformations

# Household number and city are will be denominators for food sovereignty analysis.
# Exchange rates for USD/TRY taken from National Bank of the Republic of Türkiye with mean of sale and offer.
# Exchange rates are from the first day and last day of the year, buy and sell rate averaged.
# Exchange rates https://www.tcmb.gov.tr/kurlar/kurlar_tr.html

data_exchangerates = pd.DataFrame({
    'Year': [2008, 2010, 2012, 2014, 2016, 2018, 2020, 2022, 2024],
    'Exchange_Parity_USDTRY': [1.3449, 1.5079, 1.8316, 2.2513, 3.2335, 4.5271, 6.6895, 16.0746, 32.5031]})

data_official_Türkiye = data_official_Türkiye.merge(
    data_exchangerates, on="Year", how="left")


data_official_Türkiye["Mean_Household_Count"] = data_official_Türkiye["Population_Total"] / data_official_Türkiye["Mean_Household_Size"]
data_official_Türkiye["Landuse_Harvest_km2"] = data_official_Türkiye["Landuse_Harvest_Hectare"] * 0.01
data_official_Türkiye["Greenhouse_Agro_LandUse_km2"] = data_official_Türkiye["Greenhouse_Agro_LandUse_Decare"] * 0.001
data_official_Türkiye["LandUse_Sowed_km2"] = data_official_Türkiye["LandUse_Sowed_Hectare"] * 0.01
data_official_Türkiye["Agriculture_LandUse_Fallow_km2"] = data_official_Türkiye["Agriculture_LandUse_Fallow_Hectare"] * 0.01
data_official_Türkiye["Agriculture_LandUse_Vegetables_km2"] = data_official_Türkiye["Agriculture_LandUse_Vegetables_Hectare"] * 0.01
data_official_Türkiye["Agriculture_Longterm_Crops_km2"] = data_official_Türkiye["Agriculture_Longterm_Crops_Hectare"] * 0.01


data_official_Türkiye = data_official_Türkiye.drop(columns=["Population_Total", "Landuse_Harvest_Hectare", "Greenhouse_Agro_LandUse_Decare",
                                                            "LandUse_Sowed_Hectare", "Agriculture_LandUse_Fallow_Hectare", "Agriculture_LandUse_Vegetables_Hectare",
                                                            "Agriculture_Longterm_Crops_Hectare"])  # Mean_Household_Size kept for now -- needed by the extended-indicators cell
```

## Extended Indicators — Economic Value, Water, Energy (partial year coverage)

TÜİK has not published municipal-level agricultural production *value* (crop/livestock/animal
products, in TL) for 2022 or 2024 yet, and municipal water/agricultural-electricity stats are
missing for 2024. Rather than carry `NaN`s for those years into the main FSOI panel, these
indicators are split into `data_official_Türkiye_extended` and analysed on their own below —
this maps onto the Economic conditions / Ecological conditions categories from the literature
review (see `variables_agrolife_econometrics.md`).

```python
# Extended indicators: economic value (crop/livestock/animal products, USD) + water + agricultural electricity.
# These have 2022/2024 gaps (see markdown above), so they're kept separate from the main FSOI panel.

data_official_Türkiye_extended = data_official_Türkiye[[
    "Year", "Location_Name", "Treated", "Treated_Label", "Area_km2", "Mean_Household_Count", "Mean_Household_Size",
    "Agricultural_Production_1000TL", "Agricultural_Production_Livestock_1000TL", "Agricultural_Production_AnimalProducts_1000TL",
    "Agricultural_Production_PerCapita_Crops_TL", "Agricultural_Production_PerCapita_Livestock_TL",
    "Exchange_Parity_USDTRY",
    "Water_Drainage_1000m3PerYear", "Water_Refined_1000m3PerYear",
    "Electric_Energy_Use_Agriculture_MWh"
]].copy()

# Currency conversion
data_official_Türkiye_extended["Agricultural_Production_1000USD"] = data_official_Türkiye_extended["Agricultural_Production_1000TL"] / data_official_Türkiye_extended["Exchange_Parity_USDTRY"]
data_official_Türkiye_extended["Agricultural_Production_Livestock_1000USD"] = data_official_Türkiye_extended["Agricultural_Production_Livestock_1000TL"] / data_official_Türkiye_extended["Exchange_Parity_USDTRY"]
data_official_Türkiye_extended["Agricultural_Production_AnimalProducts_1000USD"] = data_official_Türkiye_extended["Agricultural_Production_AnimalProducts_1000TL"] / data_official_Türkiye_extended["Exchange_Parity_USDTRY"]

# Already per-capita, so USD conversion only (no per-Area / per-Household division)
data_official_Türkiye_extended["Agricultural_Production_PerCapita_Crops_USD"] = data_official_Türkiye_extended["Agricultural_Production_PerCapita_Crops_TL"] / data_official_Türkiye_extended["Exchange_Parity_USDTRY"]
data_official_Türkiye_extended["Agricultural_Production_PerCapita_Livestock_USD"] = data_official_Türkiye_extended["Agricultural_Production_PerCapita_Livestock_TL"] / data_official_Türkiye_extended["Exchange_Parity_USDTRY"]

# Per-household from per-capita: PerHousehold = PerCapita x Mean_Household_Size, since
# Mean_Household_Count = Population_Total / Mean_Household_Size by construction (see
# "Transformations" cell), so PerCapita x Mean_Household_Size = Total / Mean_Household_Count.
# This is a distinct derivation from agro_prod_1000USD_perHousehold (Total / Mean_Household_Count
# directly) -- TÜİK publishes the per-capita figure separately from the total, so the two won't
# be identical. Kept as a separate variable for now, pending a redundancy re-check (2026-09-11).
data_official_Türkiye_extended["Agricultural_Production_PerCapita_Crops_USD_perHousehold"] = data_official_Türkiye_extended["Agricultural_Production_PerCapita_Crops_USD"] * data_official_Türkiye_extended["Mean_Household_Size"]
data_official_Türkiye_extended["Agricultural_Production_PerCapita_Livestock_USD_perHousehold"] = data_official_Türkiye_extended["Agricultural_Production_PerCapita_Livestock_USD"] * data_official_Türkiye_extended["Mean_Household_Size"]

# Per-Area / Per-Household indicators
data_official_Türkiye_extended["agro_prod_1000USD_perArea"] = data_official_Türkiye_extended["Agricultural_Production_1000USD"] / data_official_Türkiye_extended["Area_km2"]
data_official_Türkiye_extended["agro_prod_1000USD_perHousehold"] = data_official_Türkiye_extended["Agricultural_Production_1000USD"] / data_official_Türkiye_extended["Mean_Household_Count"]

data_official_Türkiye_extended["agro_livestock_1000USD_perArea"] = data_official_Türkiye_extended["Agricultural_Production_Livestock_1000USD"] / data_official_Türkiye_extended["Area_km2"]
data_official_Türkiye_extended["agro_livestock_1000USD_perHousehold"] = data_official_Türkiye_extended["Agricultural_Production_Livestock_1000USD"] / data_official_Türkiye_extended["Mean_Household_Count"]

data_official_Türkiye_extended["agro_animalproducts_1000USD_perArea"] = data_official_Türkiye_extended["Agricultural_Production_AnimalProducts_1000USD"] / data_official_Türkiye_extended["Area_km2"]
data_official_Türkiye_extended["agro_animalproducts_1000USD_perHousehold"] = data_official_Türkiye_extended["Agricultural_Production_AnimalProducts_1000USD"] / data_official_Türkiye_extended["Mean_Household_Count"]

data_official_Türkiye_extended["water_drainage_perArea"] = data_official_Türkiye_extended["Water_Drainage_1000m3PerYear"] / data_official_Türkiye_extended["Area_km2"]
data_official_Türkiye_extended["water_drainage_perHousehold"] = data_official_Türkiye_extended["Water_Drainage_1000m3PerYear"] / data_official_Türkiye_extended["Mean_Household_Count"]

data_official_Türkiye_extended["water_refined_perArea"] = data_official_Türkiye_extended["Water_Refined_1000m3PerYear"] / data_official_Türkiye_extended["Area_km2"]
data_official_Türkiye_extended["water_refined_perHousehold"] = data_official_Türkiye_extended["Water_Refined_1000m3PerYear"] / data_official_Türkiye_extended["Mean_Household_Count"]

data_official_Türkiye_extended["electricity_agriculture_mwh_perArea"] = data_official_Türkiye_extended["Electric_Energy_Use_Agriculture_MWh"] / data_official_Türkiye_extended["Area_km2"]
data_official_Türkiye_extended["electricity_agriculture_mwh_perHousehold"] = data_official_Türkiye_extended["Electric_Energy_Use_Agriculture_MWh"] / data_official_Türkiye_extended["Mean_Household_Count"]

data_official_Türkiye_extended = data_official_Türkiye_extended.drop(columns=[
    "Agricultural_Production_1000TL", "Agricultural_Production_Livestock_1000TL", "Agricultural_Production_AnimalProducts_1000TL",
    "Agricultural_Production_PerCapita_Crops_TL", "Agricultural_Production_PerCapita_Livestock_TL",
    "Exchange_Parity_USDTRY", "Mean_Household_Size"])


# Drop these raw source columns from the main FSOI panel now that they live in the extended dataframe
data_official_Türkiye = data_official_Türkiye.drop(columns=[
    "Agricultural_Production_1000TL", "Agricultural_Production_Livestock_1000TL", "Agricultural_Production_AnimalProducts_1000TL",
    "Agricultural_Production_PerCapita_Crops_TL", "Agricultural_Production_PerCapita_Livestock_TL",
    "Exchange_Parity_USDTRY", "Water_Drainage_1000m3PerYear", "Water_Refined_1000m3PerYear",
    "Electric_Energy_Use_Agriculture_MWh", "Electric_Energy_Use_Total_MWh", "Mean_Household_Size"])

data_official_Türkiye_extended.info()
```

## New Data — Fertilizer Use (energy category, cost-framed)

`TOB_fertilizer_cities.xlsx` (Ministry of Agriculture and Forestry) — total fertilizer use
per city per year, in tons, 2000-2025, zero missing values. Filtered to this panel's analysis
years and merged into `data_official_Türkiye` (full coverage, no TÜİK publication gap, so it
belongs in main, not extended). Folded into the **energy** category as a **cost** indicator
(input-intensity), per Yilmaz (2025)'s Entropy-TOPSIS precedent treating Fertilizer Intensity
as a cost criterion — same cost/burden framing decided for water, waste, and the rest of
energy (2026-09-11).

```python
# New data source: TOB_fertilizer_cities.xlsx (Ministry of Agriculture and Forestry).
# Total fertilizer use per city per year (tons) -- folded into the "energy" category
# as a cost/negative indicator (input-intensity), per Yilmaz (2025)'s Entropy-TOPSIS
# precedent treating Fertilizer Intensity as a cost criterion.

fertilizer_raw = pd.read_excel("TOB_fertilizer_cities.xlsx", sheet_name="BİTKİ BESİN MADDESİ TÜKETİMİ", header=1)
fertilizer_raw.columns = [c.strip() for c in fertilizer_raw.columns]

analysis_years = [2008, 2010, 2012, 2014, 2016, 2018, 2020, 2022, 2024]
fertilizer_raw = fertilizer_raw[fertilizer_raw["YIL"].isin(analysis_years)].copy()

# Turkish-aware case normalization -- Python's default .title()/.upper()/.lower() mishandle
# Turkish İ/I/ı/i (e.g. "I".lower() -> "i" in Python, but should be "ı" in Turkish), so this
# file's ALL-CAPS city names ("AFYONKARAHİSAR", "ADIYAMAN") need explicit handling to match
# Location_Name's Title-case Turkish spelling ("Afyonkarahisar", "Adıyaman").
TR_UPPER_TO_LOWER = str.maketrans({'İ': 'i', 'I': 'ı'})

def tr_title(s):
    s2 = s.strip().translate(TR_UPPER_TO_LOWER).lower()
    words = []
    for w in s2.split():
        first, rest = w[0], w[1:]
        first_cap = 'İ' if first == 'i' else ('I' if first == 'ı' else first.upper())
        words.append(first_cap + rest)
    return " ".join(words)

fertilizer_raw["Location_Name"] = fertilizer_raw["İL"].apply(tr_title)

# Validate every fertilizer city name matches a real panel city before merging -- fail loudly
# rather than silently dropping/mismatching a city. (Türkiye aggregate row is expected to be
# absent from this file -- not asserted here since it's not in the "unmatched" direction.)
unmatched = set(fertilizer_raw["Location_Name"]) - set(data_official_Türkiye["Location_Name"])
assert not unmatched, f"Unmatched fertilizer city names: {unmatched}"
assert not fertilizer_raw.duplicated(subset=["YIL", "Location_Name"]).any(), "Duplicate Year/Location_Name in fertilizer data"

fertilizer_raw = fertilizer_raw.rename(columns={"YIL": "Year", "TOPLAM BİTKİ BESİN MADDESİ": "Fertilizer_Use_Total_Ton"})
fertilizer_raw = fertilizer_raw[["Year", "Location_Name", "Fertilizer_Use_Total_Ton"]]

data_official_Türkiye = data_official_Türkiye.merge(fertilizer_raw, on=["Year", "Location_Name"], how="left")

# Energy category: fertilizer use, cost-framed (Yilmaz 2025) -- perArea / perHousehold.
data_official_Türkiye["fertilizer_use_perArea"] = data_official_Türkiye["Fertilizer_Use_Total_Ton"] / data_official_Türkiye["Area_km2"]
data_official_Türkiye["fertilizer_use_perHousehold"] = data_official_Türkiye["Fertilizer_Use_Total_Ton"] / data_official_Türkiye["Mean_Household_Count"]

data_official_Türkiye = data_official_Türkiye.drop(columns=["Fertilizer_Use_Total_Ton"])

data_official_Türkiye[["Year", "Location_Name", "fertilizer_use_perArea", "fertilizer_use_perHousehold"]].head(10)
```

```python
# How redundant is each _perArea indicator with its _perHousehold counterpart?
def perarea_household_pairs(df):
    pairs = []
    for col in df.columns:
        if col.endswith("_perArea"):
            counterpart = col[: -len("_perArea")] + "_perHousehold"
            if counterpart in df.columns:
                pairs.append((col, counterpart))
    return pairs

rows = []
for label, df in [("data_official_Türkiye", data_official_Türkiye), ("data_official_Türkiye_extended", data_official_Türkiye_extended)]:
    for col_area, col_hh in perarea_household_pairs(df):
        rows.append({"dataframe": label, "perArea": col_area, "perHousehold": col_hh, "correlation": df[col_area].corr(df[col_hh])})

perarea_household_corr = pd.DataFrame(rows).sort_values("correlation")
perarea_household_corr
```

### Collapsing correlated indicator clusters into sub-indices

**Pending Orhan's confirmation (2026-09-13) — see `agent_note_econometrics_FSOI.md` for the
full correlation table before treating this as settled.**

Per the redundancy map above and the 2026-09-11 decision to use a simple mean of standardized
values (not PCA/factor — components are hard to name/justify in a thesis, and factor analysis
is near-degenerate on 2-variable clusters), some collapse candidates below are combined into one
column per cluster, replacing their constituent columns.

**perArea and perHousehold are two distinct, deliberately-kept-separate tracks and are never
blended into each other.** A cluster only ever combines two *different source variables* within
the same track (e.g. `landuse_harvested_perArea` with `landuse_sowed_perArea` — both perArea).
An earlier version of this section incorrectly collapsed `electricity_agriculture_mwh_perArea`
with its own `_perHousehold` counterpart — that has been reverted; both columns are kept
separate, uncollapsed, below.

**Standardization method:** each constituent column is independently min-max scaled to [0, 1]
(pooled across all rows, including the `Türkiye` aggregate row, matching how the correlation/
redundancy checks above already treat the panel) before the columns are averaged. Min-max, not
z-score, is used here to keep the combined value non-negative and bounded to [0, 1] — this is
only a within-cluster standardization to make otherwise incomparable units (e.g. production
tons vs. land area) averageable; the resulting combined column still goes through the
pipeline's later `log1p` + winsorized min-max normalization (pooled + per-year) like any other
indicator once benefit/cost framing is applied.

**Land-use and greenhouse clusters (main panel)** each produce both a perArea and a perHousehold
combined column, since both tracks show the same near-duplicate correlation for these two
clusters.

**Water cluster (extended panel)** collapses only the perArea forms (`water_drainage_perArea` ≈
`water_refined_perArea`, r=0.99) — their perHousehold counterparts don't show this correlation
per the redundancy map above (see the note for the exact number), so they're left as two
separate, uncollapsed columns rather than forced into symmetry with the perArea track.

**Electricity cluster is NOT collapsed** (reverted, see above) — `electricity_agriculture_mwh_perArea`
and `electricity_agriculture_mwh_perHousehold` remain two separate columns, since collapsing
them would blend the perArea/perHousehold distinction rather than combine two distinct
variables.

```python
# Simple-mean-of-standardized-values collapse for correlated clusters identified in the
# redundancy map above (2026-09-11 decision: simple mean, not PCA/factor -- see markdown).
#
# IMPORTANT: perArea and perHousehold are two distinct, deliberately-kept-separate tracks
# throughout this pipeline -- a cluster combines two *different source variables* within the
# same track (e.g. harvested-land-perArea with sowed-land-perArea), never a variable's own
# perArea form with its own perHousehold form. (An earlier version of this cell blended
# electricity_agriculture_mwh_perArea with its own _perHousehold counterpart -- that was wrong
# and has been reverted; both stay as separate, uncollapsed columns below. Flagged by Orhan,
# 2026-09-13 -- see agent_note_econometrics_FSOI.md for the corrected variable list, pending
# his confirmation before any further clustering.)

def minmax_scale(series):
    return (series - series.min()) / (series.max() - series.min())

def collapse_cluster(df, cols, new_col):
    """Combine `cols` into `new_col` as the mean of each column's own pooled min-max scaling,
    then drop the constituent columns. `cols` must all belong to the same perArea/perHousehold
    track -- never mix the two."""
    scaled = pd.concat([minmax_scale(df[c]) for c in cols], axis=1)
    df[new_col] = scaled.mean(axis=1)
    return df.drop(columns=cols)


# --- Main panel (data_official_Türkiye) ---

# Land-use core: harvested ~ sowed (r = 0.99, both perArea and perHousehold) -- same land
# measured twice (agronomically, harvested and sown area are almost the same thing).
data_official_Türkiye = collapse_cluster(
    data_official_Türkiye, ["landuse_harvested_perArea", "landuse_sowed_perArea"], "landuse_core_perArea")
data_official_Türkiye = collapse_cluster(
    data_official_Türkiye, ["landuse_harvested_perHousehold", "landuse_sowed_perHousehold"], "landuse_core_perHousehold")

# Greenhouse intensity: production ~ land under greenhouse (r = 0.89-0.96) -- one latent
# "greenhouse activity" factor rather than two independent signals.
data_official_Türkiye = collapse_cluster(
    data_official_Türkiye, ["agro_greenhouse_prod_ton_perArea", "landuse_greenhouse_perArea"], "greenhouse_intensity_perArea")
data_official_Türkiye = collapse_cluster(
    data_official_Türkiye, ["agro_greenhouse_prod_ton_perHousehold", "landuse_greenhouse_perHousehold"], "greenhouse_intensity_perHousehold")

print(sorted(data_official_Türkiye.columns))


# --- Extended panel (data_official_Türkiye_extended) ---

# Water supply: drainage ~ refined, perArea form only (r = 0.99) -- perHousehold forms don't
# show this redundancy (see redundancy map above), so they're left as separate, uncollapsed
# columns (water_drainage_perHousehold, water_refined_perHousehold) rather than forced into
# symmetry with the perArea track.
data_official_Türkiye_extended = collapse_cluster(
    data_official_Türkiye_extended, ["water_drainage_perArea", "water_refined_perArea"], "water_supply_perArea")

# Electricity: perArea and perHousehold are NOT collapsed here (reverted, see note above) --
# electricity_agriculture_mwh_perArea and electricity_agriculture_mwh_perHousehold stay as two
# separate columns, unchanged from the main transformations cell.

print(sorted(data_official_Türkiye_extended.columns))
```
