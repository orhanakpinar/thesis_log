```python
# Deleting town area values, keeping only city area values

data_HGM = data_HGM[
    data_HGM["İL VE İLÇE YÜZ ÖLÇÜMLERİ"].isna() &
    data_HGM["Unnamed: 1"].isna() &
    data_HGM["Unnamed: 2"].notna()]

data_HGM.reset_index(inplace=True, drop=True)
print(data_HGM)
```

```python
import unicodedata

# Correction of city name values for Türkiye-HGM data

data_HGM = data_HGM.rename(columns={"Unnamed: 2" : "Location_Name", 
                        "Unnamed: 3" : "Area_km2"})

data_HGM = data_HGM.drop(columns=["İL VE İLÇE YÜZ ÖLÇÜMLERİ", "Unnamed: 1"])
data_HGM.drop(index=0, axis=0, inplace=True)
data_HGM.reset_index(inplace=True, drop=True)


# Capitalize city names while preserving Turkish encoding
def normalize_turkish(text):
    if not isinstance(text, str):
        return text
    # Normalize Unicode composition (fixes dotted İ etc.)
    text = unicodedata.normalize("NFKC", text.strip())
    # Replace locale-specific characters before capitalization
    text = text.translate(str.maketrans("IİŞĞÜÇÖ", "ıişğüçö")).capitalize()
    return text

# Apply to city names
data_HGM["Location_Name"] = data_HGM["Location_Name"].apply(normalize_turkish)


print(data_HGM["Location_Name"].unique())
# For Hakkari, İstanbul, and İzmir
data_HGM["Location_Name"] = data_HGM["Location_Name"].str.replace("Hakkâri", "Hakkari")
data_HGM["Location_Name"] = data_HGM["Location_Name"].str.replace("Istanbul", "İstanbul")
data_HGM["Location_Name"] = data_HGM["Location_Name"].str.replace("Izmir", "İzmir")
print(data_HGM["Location_Name"].unique())

```

```python
data_official_Türkiye = data_TÜİK.merge(
    data_HGM.assign(Location_lower=data_HGM["Location_Name"].str.lower())[["Location_lower", "Area_km2"]],
    left_on=data_TÜİK["Location_Name"].str.lower(),
    right_on="Location_lower",
    how="left"
).drop(columns="Location_lower")
data_official_Türkiye["Area_km2"] = data_official_Türkiye["Area_km2"].astype("float64")
```

```python
# Control empty area values

missing_area = data_official_Türkiye[data_official_Türkiye["Area_km2"].isna()]
print(missing_area["Location_Name"].unique())

# Because that I replaced the city encodings, only Türkiye was left, picked a reference year 2008 

total_area = data_official_Türkiye.loc[
    (data_official_Türkiye["Year"] == 2008) & (data_official_Türkiye["Location_Name"] != "Türkiye"), "Area_km2"].sum()

# Assign total to Türkiye
data_official_Türkiye.loc[data_official_Türkiye["Location_Name"] == "Türkiye", "Area_km2"] = total_area

missing_area = data_official_Türkiye[data_official_Türkiye["Area_km2"].isna()]
print(missing_area["Location_Name"].unique())


```

```python
# Transformations

# Household number and city are will be denominators for food sovereignty analysis.
# Exchange rates for USD/TRY taken from National Bank of the Republic of Türkiye with mean of sale and offer.
# Exchange rates are from the first day and last day of the year, buy and sell rate averaged.
# Exchange rates https://www.tcmb.gov.tr/kurlar/kurlar_tr.html

data_exchangerates = pd.DataFrame({
    'Year': [2008, 2010, 2012, 2014, 2016, 2018, 2020, 2022, 2024],
    'Exchange_Parity_USDTRY': [1.3449, 1.5079, 1.8316, 2.2513, 3.2335, 4.5271, 6.6895, 16.0746, 32.5031]})

data_official_Türkiye = data_official_Türkiye.merge(
    data_exchangerates, on="Year", how="left")


data_official_Türkiye["Mean_Household_Count"] = data_official_Türkiye["Population_Total"] / data_official_Türkiye["Mean_Household_Size"]
data_official_Türkiye["Landuse_Harvest_km2"] = data_official_Türkiye["Landuse_Harvest_Hectare"] * 0.01
data_official_Türkiye["Greenhouse_Agro_LandUse_km2"] = data_official_Türkiye["Greenhouse_Agro_LandUse_Decare"] * 0.001
data_official_Türkiye["LandUse_Sowed_km2"] = data_official_Türkiye["LandUse_Sowed_Hectare"] * 0.01
data_official_Türkiye["Agriculture_LandUse_Fallow_km2"] = data_official_Türkiye["Agriculture_LandUse_Fallow_Hectare"] * 0.01
data_official_Türkiye["Agriculture_LandUse_Vegetables_km2"] = data_official_Türkiye["Agriculture_LandUse_Vegetables_Hectare"] * 0.01
data_official_Türkiye["Agriculture_Longterm_Crops_km2"] = data_official_Türkiye["Agriculture_Longterm_Crops_Hectare"] * 0.01


data_official_Türkiye = data_official_Türkiye.drop(columns=["Landuse_Harvest_Hectare", "Greenhouse_Agro_LandUse_Decare",
                                                            "LandUse_Sowed_Hectare", "Agriculture_LandUse_Fallow_Hectare", "Agriculture_LandUse_Vegetables_Hectare",
                                                            "Agriculture_Longterm_Crops_Hectare"])
# Population_Total is NOT dropped here (unlike the columns above) -- the leftover-column
# denomination step right below still needs it directly; it drops it itself once done
# (2026-09-17, moved up from where it used to be reconstructed further downstream).
# Mean_Household_Size is also kept for now -- needed by both the leftover-column step below
# and the extended-indicators cell after it.
```

## Leftover Columns — perArea/perHousehold Denomination (resolved 2026-09-17/18)

Four raw TÜİK columns were carried through the pipeline without ever being folded into the
perArea/perHousehold indicator set: `Population_Density_PeoplePerKm2`, `Total_Agricultural_Production_Ton`,
`Water_Drainage_LitrePerPersonPerDay`, `Waste_Collected_KgPerPersonPerDay`. This step runs here,
right after `Mean_Household_Count`/`Population_Total`/`Area_km2` are all available and before
anything downstream needs them — a pretransformation step, rather than threading these columns
through several intervening cells (an earlier version sat much further down and had to keep
`Mean_Household_Size` alive past its original drop point to reach it; this placement avoids that
and uses `Population_Total` directly instead of reconstructing it).

- **`Population_Density_PeoplePerKm2` — dropped, not promoted.** 0.995-correlated with
  `wasteCollected_1000ton_perArea` (see redundancy map further below) — a near-duplicate, not an
  independent signal. Confirmed unrelated to `Mean_Household_Count`'s construction (that uses
  `Population_Total`, not this column, despite the similarly-themed name).
- **`Total_Agricultural_Production_Ton`** is already a raw city-year total (crop production,
  tons) → standard `Total / Area_km2`, `Total / Mean_Household_Count` split.
- **`Waste_Collected_KgPerPersonPerDay`** is a per-person rate, not a total — scaled up to a
  city-day total (`rate x Population_Total`) and then split into the standard
  perArea/perHousehold pair.
- **`Water_Drainage_LitrePerPersonPerDay` is deliberately NOT used** (2026-09-19). TÜİK
  computes it per person *in municipalities*, and Law 6360 moved that population base for
  treated cities in 2014, so the series carries the reform in its own denominator (see
  Diagnostics). Water is represented by the extended panel's annual-total drawn series
  instead — which costs no coverage, since both series are missing exactly 2024.

**Naming** (settled 2026-09-18): `waste_collected_kg_daily_perArea`/`perHousehold`.
"PerPerson" is dropped — it describes the *source* unit, not the derived indicator, once the
rate has been scaled to a population total — but the unit (kg) and the daily grain stay
explicit. Land-use indicators carry `_km2` for the same reason; the only names without a unit
suffix are the collapsed cluster indices further below, which are unitless by construction.



```python
data_official_Türkiye = data_official_Türkiye.drop(columns=["Population_Density_PeoplePerKm2"])

# Turkish comma-decimal string; at least one value also has a leading non-breaking space
# (U+00A0, e.g. '\xa01,15') -- .str.strip() before the comma/dot replace handles it (found by
# thesis_log_econometrics_agent_old_01, 2026-09-13; a plain .str.replace(',', '.').astype(float)
# would raise on that row).
data_official_Türkiye["Waste_Collected_KgPerPersonPerDay"] = (
    data_official_Türkiye["Waste_Collected_KgPerPersonPerDay"]
    .str.strip()
    .str.replace(",", ".", regex=False)
    .astype(float)
)

# Total_Agricultural_Production_Ton: already a raw total -- standard Total/Area, Total/Household split.
data_official_Türkiye["total_agro_production_ton_perArea"] = data_official_Türkiye["Total_Agricultural_Production_Ton"] / data_official_Türkiye["Area_km2"]
data_official_Türkiye["total_agro_production_ton_perHousehold"] = data_official_Türkiye["Total_Agricultural_Production_Ton"] / data_official_Türkiye["Mean_Household_Count"]

# Water_Drainage_LitrePerPersonPerDay / Waste_Collected_KgPerPersonPerDay: per-person rates, not
# totals -- scale up to a city-day total using Population_Total directly (still available here,
# not yet dropped), then split into the standard perArea/perHousehold pair.
#
# Naming: "PerPerson" is dropped (it describes the source unit, not the derived indicator, once
# scaled to a population total), but the unit (litre/kg) and the daily grain stay explicit, per
# Orhan 2026-09-18 -- the same way the rest of the pipeline keeps units visible. "drainage" here
# is the corrected translation of "çekilen" (drawn), not "arıtılan" (treated) -- see the source
# rename cell near the top.
# Neither per-person rate is turned into an indicator any more (2026-09-19, per Orhan).
#
# WATER: TUIK computes its per-person water rate per person IN MUNICIPALITIES, and Law 6360 moved
# that population base for treated cities in 2014 (see Diagnostics), so the series carries the
# reform in its own denominator. Water is represented by the extended panel's annual-total drawn
# series instead -- which costs no coverage, both series being missing exactly 2024. The main panel
# therefore holds no water column, and is gap-free in exchange.
#
# WASTE: the annual total (Waste_Collected_1000Ton) already supplies waste, and the daily-rate
# derivation was 0.999-correlated with it in perArea form. Dropping both its perArea and
# perHousehold forms together keeps the perArea/perHousehold symmetry rule intact.
#
# Both raw columns are still cleaned above and kept in `diagnostic_raw` for the Diagnostics section.

# Snapshot of the raw columns the Diagnostics section needs, taken here while they still exist.
# Keeping this means the diagnostics below COMPUTE their numbers rather than restating them.
diagnostic_raw = data_official_Türkiye[[
    "Year", "Location_Name", "Treated_Label", "Population_Total", "Area_km2",
    "Mean_Household_Count", "Mean_Household_Size",
    "Water_Drainage_1000m3PerYear", "Water_Refined_1000m3PerYear", "Waste_Collected_1000Ton",
    "Water_Drainage_LitrePerPersonPerDay", "Waste_Collected_KgPerPersonPerDay",
    "Landuse_Harvest_km2", "Total_Agricultural_Production_Ton", "Greenhouse_Agro_Production_Ton",
]].copy()

data_official_Türkiye = data_official_Türkiye.drop(columns=[
    "Population_Total", "Total_Agricultural_Production_Ton", "Water_Drainage_LitrePerPersonPerDay", "Waste_Collected_KgPerPersonPerDay"])

data_official_Türkiye.columns
```

## Extended Indicators — Economic Value, Water, Energy (partial year coverage)

TÜİK has not published municipal-level agricultural production *value* (crop/livestock/animal
products, in TL) for 2022 or 2024 yet, and municipal water/agricultural-electricity stats are
missing for 2024. Rather than carry `NaN`s for those years into the main FSOI panel, these
indicators are split into `data_official_Türkiye_extended` and analysed on their own below —
this maps onto the Economic conditions / Ecological conditions categories from the literature
review (see `variables_agrolife_econometrics.md`).

```python
# Extended indicators: economic value (crop/livestock/animal products, USD) + water + agricultural electricity.
# These have 2022/2024 gaps (see markdown above), so they're kept separate from the main FSOI panel.
#
# Electric_Energy_Use_Total_MWh is carried into the extended panel as a COVARIATE, not an FSOI
# indicator (added 2026-09-19): total-minus-agricultural electricity is the only available proxy
# for industrial/non-agricultural activity in this dataset, and the diagnostics section further
# down uses it to test whether industrialization explains the perArea/perHousehold divergence.
# It is not part of any of the six categories.

data_official_Türkiye_extended = data_official_Türkiye[[
    "Year", "Location_Name", "Treated", "Treated_Label", "Area_km2", "Mean_Household_Count", "Mean_Household_Size",
    "Agricultural_Production_Crop_1000TL", "Agricultural_Production_Livestock_1000TL", "Agricultural_Production_AnimalProducts_1000TL",
    "Agricultural_Production_PerCapita_Crops_TL", "Agricultural_Production_PerCapita_Livestock_TL",
    "Exchange_Parity_USDTRY",
    "Water_Drainage_1000m3PerYear", "Water_Refined_1000m3PerYear",
    "Electric_Energy_Use_Agriculture_MWh", "Electric_Energy_Use_Total_MWh"
]].copy()

# Currency conversion
data_official_Türkiye_extended["Agricultural_Production_Crop_1000USD"] = data_official_Türkiye_extended["Agricultural_Production_Crop_1000TL"] / data_official_Türkiye_extended["Exchange_Parity_USDTRY"]
data_official_Türkiye_extended["Agricultural_Production_Livestock_1000USD"] = data_official_Türkiye_extended["Agricultural_Production_Livestock_1000TL"] / data_official_Türkiye_extended["Exchange_Parity_USDTRY"]
data_official_Türkiye_extended["Agricultural_Production_AnimalProducts_1000USD"] = data_official_Türkiye_extended["Agricultural_Production_AnimalProducts_1000TL"] / data_official_Türkiye_extended["Exchange_Parity_USDTRY"]

# Already per-capita, so USD conversion only (no per-Area / per-Household division)
data_official_Türkiye_extended["Agricultural_Production_PerCapita_Crops_USD"] = data_official_Türkiye_extended["Agricultural_Production_PerCapita_Crops_TL"] / data_official_Türkiye_extended["Exchange_Parity_USDTRY"]
data_official_Türkiye_extended["Agricultural_Production_PerCapita_Livestock_USD"] = data_official_Türkiye_extended["Agricultural_Production_PerCapita_Livestock_TL"] / data_official_Türkiye_extended["Exchange_Parity_USDTRY"]

# Per-household from per-capita: PerHousehold = PerCapita x Mean_Household_Size, since
# Mean_Household_Count = Population_Total / Mean_Household_Size by construction (see
# "Transformations" cell), so PerCapita x Mean_Household_Size = Total / Mean_Household_Count.
# Used here only as a cross-check against agro_crop_1000USD_perHousehold (Total / Mean_Household_Count
# directly) -- confirmed to agree to 7 decimal places (see redundancy-cleanup cell below), so this
# derived pair is dropped there, not kept as an independent indicator.
data_official_Türkiye_extended["Agricultural_Production_PerCapita_Crops_USD_perHousehold"] = data_official_Türkiye_extended["Agricultural_Production_PerCapita_Crops_USD"] * data_official_Türkiye_extended["Mean_Household_Size"]
data_official_Türkiye_extended["Agricultural_Production_PerCapita_Livestock_USD_perHousehold"] = data_official_Türkiye_extended["Agricultural_Production_PerCapita_Livestock_USD"] * data_official_Türkiye_extended["Mean_Household_Size"]

# Per-Area / Per-Household indicators
data_official_Türkiye_extended["agro_crop_1000USD_perArea"] = data_official_Türkiye_extended["Agricultural_Production_Crop_1000USD"] / data_official_Türkiye_extended["Area_km2"]
data_official_Türkiye_extended["agro_crop_1000USD_perHousehold"] = data_official_Türkiye_extended["Agricultural_Production_Crop_1000USD"] / data_official_Türkiye_extended["Mean_Household_Count"]

data_official_Türkiye_extended["agro_livestock_1000USD_perArea"] = data_official_Türkiye_extended["Agricultural_Production_Livestock_1000USD"] / data_official_Türkiye_extended["Area_km2"]
data_official_Türkiye_extended["agro_livestock_1000USD_perHousehold"] = data_official_Türkiye_extended["Agricultural_Production_Livestock_1000USD"] / data_official_Türkiye_extended["Mean_Household_Count"]

data_official_Türkiye_extended["agro_animalproducts_1000USD_perArea"] = data_official_Türkiye_extended["Agricultural_Production_AnimalProducts_1000USD"] / data_official_Türkiye_extended["Area_km2"]
data_official_Türkiye_extended["agro_animalproducts_1000USD_perHousehold"] = data_official_Türkiye_extended["Agricultural_Production_AnimalProducts_1000USD"] / data_official_Türkiye_extended["Mean_Household_Count"]

data_official_Türkiye_extended["water_drainage_perArea"] = data_official_Türkiye_extended["Water_Drainage_1000m3PerYear"] / data_official_Türkiye_extended["Area_km2"]
data_official_Türkiye_extended["water_drainage_perHousehold"] = data_official_Türkiye_extended["Water_Drainage_1000m3PerYear"] / data_official_Türkiye_extended["Mean_Household_Count"]

# REFINED (arıtılan) water is deliberately NOT turned into an indicator (2026-09-19, per Orhan).
# It is dominated by whether a treatment plant exists -- a third of cities recorded exactly zero in
# 2008, falling to ~11% by 2022 -- so it measures infrastructure rollout, not water use. In a
# COST-framed water category it would also point the wrong way (more treatment = better service,
# but a cost framing would score it as worse). The raw column is kept only for the diagnostics.
# Water is therefore represented by DRAWN (çekilen) water alone, as a symmetric perArea/perHousehold
# pair -- which also resolves the old perArea/perHousehold asymmetry in this cluster.

data_official_Türkiye_extended["electricity_agriculture_mwh_perArea"] = data_official_Türkiye_extended["Electric_Energy_Use_Agriculture_MWh"] / data_official_Türkiye_extended["Area_km2"]
data_official_Türkiye_extended["electricity_agriculture_mwh_perHousehold"] = data_official_Türkiye_extended["Electric_Energy_Use_Agriculture_MWh"] / data_official_Türkiye_extended["Mean_Household_Count"]

# Covariate (not an FSOI indicator): non-agricultural electricity as an industrial/urban-activity proxy.
data_official_Türkiye_extended["nonagri_electricity_mwh_perHousehold"] = (
    data_official_Türkiye_extended["Electric_Energy_Use_Total_MWh"] - data_official_Türkiye_extended["Electric_Energy_Use_Agriculture_MWh"]
) / data_official_Türkiye_extended["Mean_Household_Count"]

# Drop the bare per-capita USD columns and their derived _perHousehold cross-check (2026-09-17,
# per Orhan): agro_crop_1000USD_perHousehold / agro_livestock_1000USD_perHousehold already carry
# this information (confirmed near-identical, r=0.9999999, below), so keeping the per-capita
# originals too would just be redundancy, not an independent indicator.
# Pairing note: per-capita is a fundamentally different normalization basis than perArea/perHousehold
# -- there is no meaningful "per-Area" version of a per-capita figure (you'd need to go back through
# total population and area, at which point you're just re-deriving agro_crop_1000USD_perArea).
# So dropping per-capita here does NOT leave a perArea/perHousehold pair with a missing side for this
# variable -- the market category's perArea+perHousehold pair for crops/livestock is fully supplied
# by agro_crop_1000USD_perArea/perHousehold and agro_livestock_1000USD_perArea/perHousehold on their
# own; per-capita was only ever a second, non-paired way of looking at the same value.
data_official_Türkiye_extended = data_official_Türkiye_extended.drop(columns=[
    "Agricultural_Production_Crop_1000TL", "Agricultural_Production_Livestock_1000TL", "Agricultural_Production_AnimalProducts_1000TL",
    "Agricultural_Production_PerCapita_Crops_TL", "Agricultural_Production_PerCapita_Livestock_TL",
    "Agricultural_Production_PerCapita_Crops_USD", "Agricultural_Production_PerCapita_Livestock_USD",
    "Agricultural_Production_PerCapita_Crops_USD_perHousehold", "Agricultural_Production_PerCapita_Livestock_USD_perHousehold",
    "Exchange_Parity_USDTRY", "Mean_Household_Size"])


# Drop these raw source columns from the main FSOI panel now that they live in the extended dataframe.
# Mean_Household_Size can be dropped here again (back to the original, simpler behavior) now that
# the leftover-column step runs earlier, right after it's created -- no more need to keep it alive
# past this cell (2026-09-17 reorg).
data_official_Türkiye = data_official_Türkiye.drop(columns=[
    "Agricultural_Production_Crop_1000TL", "Agricultural_Production_Livestock_1000TL", "Agricultural_Production_AnimalProducts_1000TL",
    "Agricultural_Production_PerCapita_Crops_TL", "Agricultural_Production_PerCapita_Livestock_TL",
    "Exchange_Parity_USDTRY", "Water_Drainage_1000m3PerYear", "Water_Refined_1000m3PerYear",
    "Electric_Energy_Use_Agriculture_MWh", "Electric_Energy_Use_Total_MWh", "Mean_Household_Size"])

data_official_Türkiye_extended.info()
```

## New Data — Fertilizer Use (energy category, cost-framed)

`TOB_fertilizer_cities.xlsx` (Ministry of Agriculture and Forestry) — total fertilizer use
per city per year, in tons, 2000-2025, zero missing values. Filtered to this panel's analysis
years and merged into `data_official_Türkiye` (full coverage, no TÜİK publication gap, so it
belongs in main, not extended). Folded into the **energy** category as a **cost** indicator
(input-intensity), per Yilmaz (2025)'s Entropy-TOPSIS precedent treating Fertilizer Intensity
as a cost criterion — same cost/burden framing decided for water, waste, and the rest of
energy (2026-09-11).

```python
# New data source: TOB_fertilizer_cities.xlsx (Ministry of Agriculture and Forestry).
# Total fertilizer use per city per year (tons) -- folded into the "energy" category
# as a cost/negative indicator (input-intensity), per Yilmaz (2025)'s Entropy-TOPSIS
# precedent treating Fertilizer Intensity as a cost criterion.

fertilizer_raw = pd.read_excel("TOB_fertilizer_cities.xlsx", sheet_name="BİTKİ BESİN MADDESİ TÜKETİMİ", header=1)
fertilizer_raw.columns = [c.strip() for c in fertilizer_raw.columns]

analysis_years = [2008, 2010, 2012, 2014, 2016, 2018, 2020, 2022, 2024]
fertilizer_raw = fertilizer_raw[fertilizer_raw["YIL"].isin(analysis_years)].copy()

# Turkish-aware case normalization -- Python's default .title()/.upper()/.lower() mishandle
# Turkish İ/I/ı/i (e.g. "I".lower() -> "i" in Python, but should be "ı" in Turkish), so this
# file's ALL-CAPS city names ("AFYONKARAHİSAR", "ADIYAMAN") need explicit handling to match
# Location_Name's Title-case Turkish spelling ("Afyonkarahisar", "Adıyaman").
TR_UPPER_TO_LOWER = str.maketrans({'İ': 'i', 'I': 'ı'})

def tr_title(s):
    s2 = s.strip().translate(TR_UPPER_TO_LOWER).lower()
    words = []
    for w in s2.split():
        first, rest = w[0], w[1:]
        first_cap = 'İ' if first == 'i' else ('I' if first == 'ı' else first.upper())
        words.append(first_cap + rest)
    return " ".join(words)

fertilizer_raw["Location_Name"] = fertilizer_raw["İL"].apply(tr_title)

# Validate every fertilizer city name matches a real panel city before merging -- fail loudly
# rather than silently dropping/mismatching a city. (Türkiye aggregate row is expected to be
# absent from this file -- it is constructed below by summing cities.)
unmatched = set(fertilizer_raw["Location_Name"]) - set(data_official_Türkiye["Location_Name"])
assert not unmatched, f"Unmatched fertilizer city names: {unmatched}"
assert not fertilizer_raw.duplicated(subset=["YIL", "Location_Name"]).any(), "Duplicate Year/Location_Name in fertilizer data"

fertilizer_raw = fertilizer_raw.rename(columns={"YIL": "Year", "TOPLAM BİTKİ BESİN MADDESİ": "Fertilizer_Use_Total_Ton"})
fertilizer_raw = fertilizer_raw[["Year", "Location_Name", "Fertilizer_Use_Total_Ton"]]

data_official_Türkiye = data_official_Türkiye.merge(fertilizer_raw, on=["Year", "Location_Name"], how="left")

# --- Gap handling (2026-09-19, per Orhan) ---
# 1. City-level gaps are treated as genuine zeros. The only affected case is Hakkari
#    (2020/2022/2024), which also records legitimate zeros in 2012/2014/2016 -- so "absent from
#    the file" reads as "no recorded fertilizer use", not "unknown". This is an imputation
#    assumption, not a fact from the source; the assertion below keeps it honest by failing if a
#    future version of the source file drops a large number of city-years, which would make
#    blanket zero-filling wrong.
is_city = data_official_Türkiye["Location_Name"] != "Türkiye"
city_gaps = int(data_official_Türkiye.loc[is_city, "Fertilizer_Use_Total_Ton"].isna().sum())
assert city_gaps <= 20, (
    f"{city_gaps} city-year fertilizer gaps -- too many to treat as genuine zeros, check the source file")
data_official_Türkiye.loc[is_city, "Fertilizer_Use_Total_Ton"] = (
    data_official_Türkiye.loc[is_city, "Fertilizer_Use_Total_Ton"].fillna(0.0))

# 2. The Türkiye aggregate row is not in the source file, so build it as the sum of the 81 cities
#    for each year -- the same construction already used for Türkiye's Area_km2 further up.
turkiye_totals = (data_official_Türkiye.loc[is_city]
                  .groupby("Year", observed=True)["Fertilizer_Use_Total_Ton"].sum())
is_turkiye = data_official_Türkiye["Location_Name"] == "Türkiye"
data_official_Türkiye.loc[is_turkiye, "Fertilizer_Use_Total_Ton"] = (
    data_official_Türkiye.loc[is_turkiye, "Year"].map(turkiye_totals).values)

assert data_official_Türkiye["Fertilizer_Use_Total_Ton"].notna().all(), "Fertilizer still has gaps after filling"

# Energy category: fertilizer use, cost-framed (Yilmaz 2025) -- perArea / perHousehold.
data_official_Türkiye["fertilizer_use_perArea"] = data_official_Türkiye["Fertilizer_Use_Total_Ton"] / data_official_Türkiye["Area_km2"]
data_official_Türkiye["fertilizer_use_perHousehold"] = data_official_Türkiye["Fertilizer_Use_Total_Ton"] / data_official_Türkiye["Mean_Household_Count"]

data_official_Türkiye = data_official_Türkiye.drop(columns=["Fertilizer_Use_Total_Ton"])

data_official_Türkiye[["Year", "Location_Name", "fertilizer_use_perArea", "fertilizer_use_perHousehold"]].head(10)
```

### Correlation / redundancy check

Groundwork before committing to an index formula: how correlated are the candidate indicators with each other, and specifically, how redundant is a `_perArea` version of an indicator with its `_perHousehold` counterpart?

```python
# How redundant is each _perArea indicator with its _perHousehold counterpart?
def perarea_household_pairs(df):
    pairs = []
    for col in df.columns:
        if col.endswith("_perArea"):
            counterpart = col[: -len("_perArea")] + "_perHousehold"
            if counterpart in df.columns:
                pairs.append((col, counterpart))
    return pairs

rows = []
for label, df in [("data_official_Türkiye", data_official_Türkiye), ("data_official_Türkiye_extended", data_official_Türkiye_extended)]:
    for col_area, col_hh in perarea_household_pairs(df):
        rows.append({"dataframe": label, "perArea": col_area, "perHousehold": col_hh, "correlation": df[col_area].corr(df[col_hh])})

perarea_household_corr = pd.DataFrame(rows).sort_values("correlation")
perarea_household_corr
```

### Detailed high-correlation-pairs table

The heatmaps and the perArea/perHousehold self-check above don't directly surface *which* variable
pairs are near-duplicates — this does, as a live, re-runnable cell so it stays accurate as
variables are added, renamed, or dropped upstream.

Raw/denominator columns (`Area_km2`, `Mean_Household_Count`, un-normalized totals) and the
non-indicator covariates are excluded from the extended-panel version — they're dominated by city
size, which would drown out genuine indicator-to-indicator redundancy.

**Read the perArea numbers with care.** As the diagnostics section below shows, `perArea` divides
by an administrative boundary that varies roughly 50-fold across cities and is unrelated to what
the indicator measures, so *any* two population-proportional quantities tend to correlate at
r > 0.95 in that form. A high perArea correlation is therefore weak evidence of genuine
redundancy; the matching perHousehold correlation is the more informative number.

```python
# Raw/denominator/covariate columns are excluded -- they are dominated by city size (or are not
# FSOI indicators at all), which would drown out genuine indicator-to-indicator redundancy.
EXTENDED_EXCLUDE = ["Area_km2", "Mean_Household_Count", "Agricultural_Production_Crop_1000USD",
                     "Agricultural_Production_Livestock_1000USD", "Agricultural_Production_AnimalProducts_1000USD",
                     "Water_Drainage_1000m3PerYear", "Water_Refined_1000m3PerYear",
                     "Electric_Energy_Use_Agriculture_MWh", "Electric_Energy_Use_Total_MWh",
                     "nonagri_electricity_mwh_perHousehold"]

def high_corr_pairs(df, exclude=(), threshold=0.7):
    numeric = df.select_dtypes(include="number").drop(columns=[c for c in exclude if c in df.columns])
    corr = numeric.corr()
    cols = corr.columns
    pairs = [
        (cols[i], cols[j], round(corr.iloc[i, j], 4))
        for i in range(len(cols)) for j in range(i + 1, len(cols))
        if abs(corr.iloc[i, j]) >= threshold
    ]
    return pd.DataFrame(pairs, columns=["variable_a", "variable_b", "correlation"]).sort_values(
        "correlation", key=abs, ascending=False).reset_index(drop=True)

main_high_corr = high_corr_pairs(data_official_Türkiye, threshold=0.7)
print(f"Main panel -- {len(main_high_corr)} indicator pairs with |r| >= 0.7:")
main_high_corr
```

```python
extended_high_corr = high_corr_pairs(data_official_Türkiye_extended, EXTENDED_EXCLUDE, threshold=0.7)
print(f"Extended panel -- {len(extended_high_corr)} indicator pairs with |r| >= 0.7 (raw/denominator columns excluded):")
extended_high_corr
```

### Diagnostics — what the perArea/perHousehold divergence actually is

Repeatedly, a `perArea` indicator correlated almost perfectly (r = 0.98–0.999) with a
*different-source* indicator measuring something only loosely related, while the matching
`perHousehold` pair did not. This section establishes why, and tests a measurement confound that
turned out to matter a great deal for the Law 6360 design.

Terminology note: the two TÜİK water series are **drawn** (*çekilen*, withdrawn into the municipal
network) and **refined** (*arıtılan*, passed through treatment). "Refined" is used rather than the
more usual English "treated" to avoid colliding with the `Treated` city-status variable.


```python
# All diagnostics below compute from `diagnostic_raw`, the snapshot taken further up before the
# raw columns are dropped. Nothing here is a restated constant.
dg = diagnostic_raw[diagnostic_raw["Location_Name"] != "Türkiye"].copy()
dg["pop_density"] = dg["Population_Total"] / dg["Area_km2"]
dg["waste_kg_rate"] = dg["Waste_Collected_KgPerPersonPerDay"].astype(float)
dg["water_perArea"] = (dg["Water_Drainage_LitrePerPersonPerDay"] * dg["Population_Total"]) / dg["Area_km2"]
dg["waste_perArea"] = dg["Waste_Collected_1000Ton"] / dg["Area_km2"]
dg["water_perHH"] = (dg["Water_Drainage_LitrePerPersonPerDay"] * dg["Population_Total"]) / dg["Mean_Household_Count"]
dg["waste_perHH"] = dg["Waste_Collected_1000Ton"] / dg["Mean_Household_Count"]
print(f"diagnostic sample: {len(dg)} city-years, {dg['Location_Name'].nunique()} cities, "
      f"{dg['Year'].min()}-{dg['Year'].max()}")
```

```python
# --- Diagnostic 1: why perArea correlates and perHousehold does not ---
# perArea = quantity / Area_km2. Area_km2 is merged from HGM on city name with no year key, so it
# is fixed geographic PROVINCE area: it does not change with Law 6360, and it is unrelated to how
# many people live there. Dividing a population-driven quantity by it leaves density standing.

n_varying = int((dg.groupby("Location_Name")["Area_km2"].nunique() > 1).sum())
print(f"Cities whose Area_km2 changes across years: {n_varying} of {dg['Location_Name'].nunique()}")
print("  -> fixed province geography, not a municipal service area.")
print()

def spread(s):
    return s.max() / s.min()

print("The size mismatch that drives everything:")
print(f"  population density   : {dg['pop_density'].min():7.1f} to {dg['pop_density'].max():7.1f} people/km2  ({spread(dg['pop_density']):5.0f}x)")
print(f"  water per person/day : {dg['Water_Drainage_LitrePerPersonPerDay'].min():7.0f} to {dg['Water_Drainage_LitrePerPersonPerDay'].max():7.0f} litres     ({spread(dg['Water_Drainage_LitrePerPersonPerDay']):5.1f}x)")
print(f"  waste per person/day : {dg['waste_kg_rate'].min():7.2f} to {dg['waste_kg_rate'].max():7.2f} kg         ({spread(dg['waste_kg_rate']):5.1f}x)")
print()
print("Density varies far more than the behaviour being measured, so 'quantity / km2' is mostly a")
print("density measurement:")
print(f"  water_perArea ~ population density : {dg['water_perArea'].corr(dg['pop_density']):+.4f}")
print(f"  waste_perArea ~ population density : {dg['waste_perArea'].corr(dg['pop_density']):+.4f}")
print()
print("Hence two unrelated municipal services look near-identical in perArea form:")
print(f"  water ~ waste, perArea      : {dg['water_perArea'].corr(dg['waste_perArea']):+.4f}")
print(f"  water ~ waste, perHousehold : {dg['water_perHH'].corr(dg['waste_perHH']):+.4f}")
print()
print("=> A high perArea correlation is WEAK evidence of redundancy. Judge it on perHousehold.")
```

```python
# --- Diagnostic 2: the Law 6360 municipal-coverage confound ---
# TUIK's per-person water column reads "BELEDIYELERDE kisi basi cekilen gunluk su miktari" --
# per person IN MUNICIPALITIES. If its denominator is municipal (not provincial) population, then
# Law 6360, which made metropolitan municipalities cover their whole province from 2014,
# mechanically moves that denominator for treated cities only.
#
# Implied municipal population = annual total drawn / (daily per-person rate * 365).
# Expressed as a share of provincial population, it should jump for new-metros at 2014.

dg["implied_muni_pop"] = (dg["Water_Drainage_1000m3PerYear"] * 1e6) / (dg["Water_Drainage_LitrePerPersonPerDay"] * 365)
dg["muni_pop_share"] = dg["implied_muni_pop"] / dg["Population_Total"]

share = dg[dg["muni_pop_share"].notna()].pivot_table(
    index="Year", columns="Treated_Label", values="muni_pop_share", aggfunc="median", observed=True)
print("Implied municipal population as a share of provincial population (median by group):")
print(share.round(3).to_string())
print()

if 2012 in share.index and 2014 in share.index:
    jump = (share.loc[2014] - share.loc[2012]).round(3)
    print("Change across the 2012 -> 2014 reform boundary:")
    for grp, val in jump.items():
        print(f"  {grp:26s} {val:+.3f}")
    print()
    new_j = jump.get("New-metropolitan (2012)")
    non_j = jump.get("Non-metropolitan")
    if new_j is not None and non_j is not None:
        print(f"New-metro minus non-metro jump: {new_j - non_j:+.3f}")
        if new_j - non_j > 0.1:
            print("=> CONFIRMED: the per-person series' own denominator moves with the reform.")
print()
print("Consequences:")
print("* Water_Drainage_LitrePerPersonPerDay carries the reform in its denominator BY CONSTRUCTION,")
print("  so its treated-vs-control gap is substantially artifact. Prefer the annual-total series.")
print("* Old-metropolitan is NOT a clean control for municipal-service variables -- Law 6360")
print("  extended existing metropolitan boundaries to the full province too.")
print("* The share above is a measurable per-city-year index of the expansion, so the confound can")
print("  potentially be CONTROLLED rather than only disclosed.")
```

```python
# --- Diagnostic 3: is refined water a usage measure or an infrastructure measure? ---
ref = dg[dg["Water_Refined_1000m3PerYear"].notna()]
z = ref.assign(is_zero=ref["Water_Refined_1000m3PerYear"] == 0).groupby("Year", observed=True)["is_zero"].agg(["sum", "count"])
z["pct_zero"] = (z["sum"] / z["count"] * 100).round(1)
print("Cities reporting EXACTLY ZERO refined (arıtılan) water:")
print(z.rename(columns={"sum": "zero_cities", "count": "cities"}).to_string())
print()

drawn = dg[dg["Water_Drainage_1000m3PerYear"].notna()]
print(f"Drawn (çekilen) water, by contrast: {int((drawn['Water_Drainage_1000m3PerYear'] == 0).sum())} "
      f"zero values out of {len(drawn)} city-years.")
print()
cover = (ref["Water_Refined_1000m3PerYear"] / ref["Water_Drainage_1000m3PerYear"]).replace([float('inf')], float('nan'))
print("National refined/drawn coverage ratio (median by year):")
print(ref.assign(cov=cover).groupby("Year", observed=True)["cov"].median().round(3).to_string())
print()
print("=> Refined water largely records WHETHER A TREATMENT PLANT EXISTS, not how much water is")
print("   used. In a COST-framed water category it would also point the wrong way (more treatment")
print("   is better service, but a cost framing scores higher values as worse).")
print("   Decision (2026-09-19): keep DRAWN, drop REFINED.")
```

```python
# --- Diagnostic 4: does perHousehold vs perCapita change who ranks well? ---
# perCapita = perHousehold / Mean_Household_Size, so any divergence comes purely from household
# size -- which is a west-east gradient. This tests Orhan's equity concern: a large-household city
# can show MORE per household while its residents have LESS per person.

print(f"Mean_Household_Size ~ population density : {dg['Mean_Household_Size'].corr(dg['pop_density']):+.4f}"
      "   (independent dimensions)")
print(f"Mean_Household_Count ~ Population_Total   : {dg['Mean_Household_Count'].corr(dg['Population_Total']):+.4f}"
      "   (household count is essentially population)")
print()

indicators = {
    "harvested land":    dg["Landuse_Harvest_km2"],
    "crop production":   dg["Total_Agricultural_Production_Ton"],
    "greenhouse output": dg["Greenhouse_Agro_Production_Ton"],
    "waste collected":   dg["Waste_Collected_1000Ton"],
    "water drawn":       dg["Water_Drainage_1000m3PerYear"],
}
print(f"  {'indicator':20s} {'Spearman(perHH, perCapita)':>28s}")
for name, total in indicators.items():
    d = pd.DataFrame({"ph": total / dg["Mean_Household_Count"],
                      "pc": total / dg["Population_Total"]}).dropna()
    print(f"  {name:20s} {d['ph'].corr(d['pc'], method='spearman'):28.3f}")
print()
print("Reading: for LAND and PRODUCTION indicators the two normalisations rank cities almost")
print("identically, so the choice is near-immaterial there -- and perHousehold is the natural unit,")
print("since the household is the farming unit. For MUNICIPAL-SERVICE indicators (water, waste) the")
print("rankings diverge markedly, and those are consumption quantities experienced per person, so")
print("that is exactly where the per-capita objection has force. Treated as a robustness check")
print("rather than a change of denominator -- see agent_note_econometrics_FSOI.md.")
```

### Diagnostic 5 — the full water trajectory, and the two institutional dates

Rather than looking only at the window around one reform, this traces drawn water per household
across the whole panel against the two dates that matter institutionally:

- **2014** — Law 6360 takes effect; metropolitan municipalities begin serving the whole province.
- **2019** — the five-year transitional waiver ends; converted villages start paying normal

Orhan's hypothesis (2026-09-19): villages previously drew water informally and for free, outside
the municipal system; the reform brought them inside it, and the waiver's end then made that water
cost money. The trajectory below is the descriptive evidence for and against that reading.


```python
# Drawn water per household, full trajectory by treatment group.
traj_src = diagnostic_raw[diagnostic_raw["Location_Name"] != "Türkiye"].copy()
traj_src["drawn_perHH"] = traj_src["Water_Drainage_1000m3PerYear"] / traj_src["Mean_Household_Count"]
traj = traj_src[traj_src["drawn_perHH"].notna()].pivot_table(
    index="Year", columns="Treated_Label", values="drawn_perHH", aggfunc="mean", observed=True)

events = {2014: "Law 6360 in force", 2020: "first obs. after waiver ends (2019)"}
print("Drawn water per household (mean by group):")
print()
header = f"  {'Year':<6}" + "".join(f"{g:>26s}" for g in traj.columns)
print(header)
for yr in traj.index:
    row = f"  {yr:<6}" + "".join(f"{traj.loc[yr, g]:26.3f}" for g in traj.columns)
    if yr in events:
        row += f"   <- {events[yr]}"
    print(row)
print()

def phase(a, b, label):
    d = traj.loc[b] - traj.loc[a]
    parts = "  ".join(f"{g.split()[0][:9]:>9s} {d[g]:+.3f}" for g in traj.columns)
    print(f"  {label:26s} {parts}")

print("Phase by phase:")
phase(2008, 2012, "pre-reform  2008->2012")
phase(2012, 2014, "reform      2012->2014")
phase(2014, 2018, "waiver in force 2014->18")
phase(2018, 2022, "after waiver 2018->2022")
print()

new, non = "New-metropolitan (2012)", "Non-metropolitan"
print("What the shape shows:")
print(f"* Pre-reform all three groups decline together ({traj.loc[2008,non]:.3f}->{traj.loc[2012,non]:.3f} non-metro,")
print(f"  {traj.loc[2008,new]:.3f}->{traj.loc[2012,new]:.3f} new-metro) -- parallel pre-trends, which is what the DiD needs.")
print(f"* At 2014 the treated group breaks upward ({traj.loc[2012,new]:.3f}->{traj.loc[2014,new]:.3f}) while the control")
print(f"  keeps falling ({traj.loc[2012,non]:.3f}->{traj.loc[2014,non]:.3f}). This is the coverage expansion.")
print(f"* It keeps rising while the waiver runs, peaking in 2018 ({traj.loc[2018,new]:.3f}).")
print(f"* After the waiver ends it falls back ({traj.loc[2018,new]:.3f}->{traj.loc[2022,new]:.3f}) while the control stays")
print(f"  flat ({traj.loc[2018,non]:.3f}->{traj.loc[2022,non]:.3f}) -- consistent with water becoming costly to villages.")
print(f"* But it does NOT return to baseline: 2022 ({traj.loc[2022,new]:.3f}) is still above 2012 ({traj.loc[2012,new]:.3f}),")
print("  while the control fell over the same span. Some of the level shift persists.")
print()
print("CAVEATS -- this is descriptive, not identification:")
print("* 2020 is also the COVID year.")
print("* The post-2018 fall is equally consistent with the initial coverage jump simply fading.")
print("* Only two observations follow the waiver's end, and the series is biennial.")
print("* Informal/private water (wells, boreholes) is invisible in both TUIK series, so the claim")
print("  that villages previously used free water cannot be verified from this data -- it is an")
print("  institutional premise, not a measurement.")
```

### Diagnostic 6 — perArea and perHousehold are contaminated in *opposite* dimensions

Diagnostic 1 showed that across cities, `perArea` is essentially population density. That is only
half the picture, and the other half points the other way. The decisive structural fact:

- **`Area_km2` is constant within a city** (0 of 81 vary across years).
- **`Mean_Household_Count` is not** — households multiply as household size shrinks.

So within a single city over time, `perArea` moves *only* because the numerator moves, while
`perHousehold` moves because both numerator and denominator move. This matters because a DiD
identifies from within-city variation over time, while a ranking of cities is cross-sectional.


```python
# Which denominator is clean in which dimension?
dd = diagnostic_raw[diagnostic_raw["Location_Name"] != "Türkiye"].copy()
dd["pop_density"] = dd["Population_Total"] / dd["Area_km2"]
dd["total"] = dd["Water_Drainage_1000m3PerYear"]
dd["perArea"] = dd["total"] / dd["Area_km2"]
dd["perHH"] = dd["total"] / dd["Mean_Household_Count"]

def within_city(a, b):
    return dd.groupby("Location_Name").apply(lambda g: g[a].corr(g[b]), include_groups=False).dropna()

wa, wh = within_city("perArea", "total"), within_city("perHH", "total")
da, dh = within_city("perArea", "pop_density"), within_city("perHH", "pop_density")

print("WITHIN a city, over time (the dimension a DiD uses):")
print(f"  perArea      ~ raw total : {wa.median():+.4f}   ({(wa > 0.99).mean():.0%} of cities above 0.99)")
print(f"  perHousehold ~ raw total : {wh.median():+.4f}   ({(wh > 0.99).mean():.0%} of cities above 0.99)")
print("  -> perArea is the raw quantity rescaled by a constant: no denominator drift at all.")
print("     perHousehold carries the shrinking-household-size trend in its denominator.")
print()
print("ACROSS cities (the dimension a ranking uses):")
print(f"  perArea      ~ density : {dd['perArea'].corr(dd['pop_density']):+.4f}")
print(f"  perHousehold ~ density : {dd['perHH'].corr(dd['pop_density']):+.4f}")
print("  -> perArea is essentially a density measurement; perHousehold is free of it.")
print()
print("Within-city correlation with density, for completeness:")
print(f"  perArea {da.median():+.4f} | perHousehold {dh.median():+.4f}")
cv = dd.groupby("Location_Name")["pop_density"].apply(lambda s: s.std() / s.mean()).median()
print(f"Within-city coefficient of variation in density: {cv:.3f} -- density separates cities, not years.")
print()
print("CONSEQUENCES FOR THE INDEX:")
print("* The two tracks are not rival estimates of one quantity. perArea measures territorial")
print("  intensity; perHousehold measures what a household has or uses. Each is clean in the")
print("  dimension the other is contaminated in.")
print("* For the DiD: city fixed effects / differencing absorb the between-city density that spoils")
print("  perArea, while they do NOT absorb perHousehold's within-city denominator drift. That makes")
print("  perArea the mechanically safer track for the causal estimate.")
print("* For ranking cities descriptively: perArea would largely rank them by density, so")
print("  perHousehold is the meaningful one there.")
print("* Averaging the two would produce a measure contaminated in BOTH dimensions and")
print("  interpretable in neither -- the same objection that ruled out PCA earlier (you could not")
print("  say in one sentence what the number means).")
```

### What the diagnostics establish (summary)

1. **`perArea` is almost entirely population density.** Province area is fixed geography, unrelated
   to population, and density varies far more across cities than the behaviour being measured. So
   *any* two population-driven quantities correlate near-perfectly in perArea form — an artifact,
   not redundancy. Judge redundancy on `perHousehold`, where water and waste are only ~0.2 related.
2. **Industrialisation does not explain it** — a non-agricultural-electricity proxy correlates
   about −0.27 with water per household, weak and the wrong sign for that story.
3. **Law 6360 changed what the municipal statistics count.** TÜİK's per-person water figure sits on
   a *municipal* population base which jumped for new-metropolitan cities in the reform year. This
   is a measurement confound sitting directly on the treatment variable, and it touches
   old-metropolitan cities too, weakening them as a control group.
4. **Refined water is an infrastructure indicator, not a usage indicator**, so it is dropped from
   the cost-framed water category; drawn water is kept as a symmetric perArea/perHousehold pair.
5. **perHousehold vs perCapita matters only for the municipal-service indicators.** Land and
   production indicators rank cities nearly identically either way; water and waste do not.

**For the six categories.** Water and waste are both municipal services sharing the confound in
(3), which is a *conceptual* reason to handle them together and a strong reason to control for
coverage — but **not** a statistical reason to merge them, since at `perHousehold` they correlate
only ~0.2. Energy is a different case (agricultural electricity and fertiliser are farm inputs, not
municipal services), so the municipal-duty argument does not extend to it.

### Collapsing correlated indicator clusters into sub-indices

**Pending Orhan's confirmation (2026-09-13) — see `agent_note_econometrics_FSOI.md` for the
full correlation table before treating this as settled.**

Per the redundancy map above and the 2026-09-11 decision to use a simple mean of standardized
values (not PCA/factor — components are hard to name/justify in a thesis, and factor analysis
is near-degenerate on 2-variable clusters), some collapse candidates below are combined into one
column per cluster, replacing their constituent columns.

**perArea and perHousehold are two distinct, deliberately-kept-separate tracks and are never
blended into each other.** A cluster only ever combines two *different source variables* within
the same track (e.g. `landuse_harvested_perArea` with `landuse_sowed_perArea` — both perArea).
An earlier version of this section incorrectly collapsed `electricity_agriculture_mwh_perArea`
with its own `_perHousehold` counterpart — that has been reverted; both columns are kept
separate, uncollapsed, below.

**Standardization method:** each constituent column is independently min-max scaled to [0, 1]
(pooled across all rows, including the `Türkiye` aggregate row, matching how the correlation/
redundancy checks above already treat the panel) before the columns are averaged. Min-max, not
z-score, is used here to keep the combined value non-negative and bounded to [0, 1] — this is
only a within-cluster standardization to make otherwise incomparable units (e.g. production
tons vs. land area) averageable; the resulting combined column still goes through the
pipeline's later `log1p` + winsorized min-max normalization (pooled + per-year) like any other
indicator once benefit/cost framing is applied.

**Land-use and greenhouse clusters (main panel)** each produce both a perArea and a perHousehold
combined column, since both tracks show the same near-duplicate correlation for these two
clusters.

**Water is no longer a cluster at all** (2026-09-19). Refined (*arıtılan*) water was dropped as an
indicator — it records whether a treatment plant exists rather than how much water is used, and it
would point the wrong way in a cost-framed category. Drawn (*çekilen*) water alone now supplies a
symmetric `water_drainage_perArea` / `water_drainage_perHousehold` pair, so there is nothing to
collapse and the old perArea/perHousehold asymmetry in this cluster is resolved. Note the r = 0.99
that originally motivated collapsing the two was a perArea artifact — see the Diagnostics section.

**Electricity cluster is NOT collapsed** (reverted, see above) — `electricity_agriculture_mwh_perArea`
and `electricity_agriculture_mwh_perHousehold` remain two separate columns, since collapsing
them would blend the perArea/perHousehold distinction rather than combine two distinct
variables.

```python
# Simple-mean-of-standardized-values collapse for correlated clusters identified in the
# redundancy map above (2026-09-11 decision: simple mean, not PCA/factor -- see markdown).
#
# IMPORTANT: perArea and perHousehold are two distinct, deliberately-kept-separate tracks
# throughout this pipeline -- a cluster combines two *different source variables* within the
# same track (e.g. harvested-land-perArea with sowed-land-perArea), never a variable's own
# perArea form with its own perHousehold form. (An earlier version of this cell blended
# electricity_agriculture_mwh_perArea with its own _perHousehold counterpart -- that was wrong
# and has been reverted; both stay as separate, uncollapsed columns below. Flagged by Orhan,
# 2026-09-13 -- see agent_note_econometrics_FSOI.md for the corrected variable list.)
#
# Note on unit suffixes: the collapsed outputs (landuse_core_*, greenhouse_intensity_*,
# water_supply_*) deliberately carry NO unit suffix, unlike their inputs. They are means of
# min-max-scaled values, so they are unitless indices by construction -- and greenhouse_intensity
# in particular averages a tonnage variable with an area variable, so no single unit would even
# be correct. Only the directly-derived indicators keep unit suffixes (_km2, _ton, _litre, _kg).

def minmax_scale(series):
    return (series - series.min()) / (series.max() - series.min())

def collapse_cluster(df, cols, new_col):
    """Combine `cols` into `new_col` as the mean of each column's own pooled min-max scaling,
    then drop the constituent columns. `cols` must all belong to the same perArea/perHousehold
    track -- never mix the two."""
    scaled = pd.concat([minmax_scale(df[c]) for c in cols], axis=1)
    df[new_col] = scaled.mean(axis=1)
    return df.drop(columns=cols)


# --- Main panel (data_official_Türkiye) ---

# Land-use core: harvested ~ sowed (r = 0.99, both perArea and perHousehold) -- same land
# measured twice (agronomically, harvested and sown area are almost the same thing).
data_official_Türkiye = collapse_cluster(
    data_official_Türkiye, ["landuse_harvested_km2_perArea", "landuse_sowed_km2_perArea"], "landuse_core_perArea")
data_official_Türkiye = collapse_cluster(
    data_official_Türkiye, ["landuse_harvested_km2_perHousehold", "landuse_sowed_km2_perHousehold"], "landuse_core_perHousehold")

# Greenhouse is deliberately NOT collapsed (2026-09-19, per Orhan). Its two inputs belong to
# DIFFERENT FSOI categories -- agro_greenhouse_prod_ton_* is production output, landuse_greenhouse_km2_*
# is land allocation -- so a combined index would belong cleanly to neither, and would also mix
# tonnes with km2 into something unitless. They stay as separate, unit-carrying indicators in their
# own categories. (The r = 0.94 that originally motivated collapsing is partly the perArea scale
# artifact documented in the Diagnostics section, so it was weaker evidence than it appeared.)

print(sorted(data_official_Türkiye.columns))


# --- Extended panel (data_official_Türkiye_extended) ---

# Water: only DRAWN water remains (refined dropped upstream, 2026-09-19), so there is nothing to
# collapse -- water_drainage_perArea / water_drainage_perHousehold are already a symmetric pair.

# Electricity: perArea and perHousehold are NOT collapsed here (reverted, see note above) --
# electricity_agriculture_mwh_perArea and electricity_agriculture_mwh_perHousehold stay as two
# separate columns, unchanged from the main transformations cell.

print(sorted(data_official_Türkiye_extended.columns))
```

# Index Construction

## Pre-normalisation profile of the final indicator set

The first thing to establish before transforming anything: what does each indicator actually look
like? This is the reference table for every normalisation choice that follows — skew drives the
`log1p` decision, the floor share drives the winsorising discussion, and the min tells us whether
`log1p` is even safe (it needs non-negative values).

Grouped by panel, since the two panels differ in year coverage and feed different categories.


```python
import numpy as np

def indicator_columns(df):
    """Indicator columns only -- excludes raw source columns, denominators and covariates."""
    return [c for c in df.columns
            if c.endswith(("_perArea", "_perHousehold"))
            and not c.startswith("nonagri_")]

# Profile of every indicator entering the index, by panel.
profile_rows = []
for panel_name, panel_df in [("main", data_official_Türkiye), ("extended", data_official_Türkiye_extended)]:
    panel_cities = panel_df[panel_df["Location_Name"] != "Türkiye"]
    for col in indicator_columns(panel_df):
        s = panel_cities[col].dropna()
        profile_rows.append({
            "panel": panel_name,
            "track": "perArea" if col.endswith("_perArea") else "perHousehold",
            "indicator": col.replace("_perArea", "").replace("_perHousehold", ""),
            "n": len(s),
            "min": s.min(),
            "median": s.median(),
            "max": s.max(),
            "max/median": (s.max() / s.median()) if s.median() else np.nan,
            "skew": s.skew(),
            "pct_zero": round((s == 0).mean() * 100, 1),
        })

indicator_profile = (pd.DataFrame(profile_rows)
                     .sort_values(["panel", "indicator", "track"])
                     .reset_index(drop=True))

print(f"{len(indicator_profile)} indicator columns "
      f"({(indicator_profile['panel'] == 'main').sum()} main, "
      f"{(indicator_profile['panel'] == 'extended').sum()} extended)")
print(f"any negative values (would make log1p unsafe)? {(indicator_profile['min'] < 0).any()}")
print(f"skew range: {indicator_profile['skew'].min():.2f} to {indicator_profile['skew'].max():.2f}")
print()
indicator_profile.style.format({
    "min": "{:.3f}", "median": "{:.3f}", "max": "{:,.1f}",
    "max/median": "{:,.1f}", "skew": "{:.2f}", "pct_zero": "{:.1f}%"})
```

## Choosing the winsorising bounds — what each option costs

`log1p` cannot fix zero-inflation, so several indicators still have most cities pressed against the
floor afterwards. Widening the winsorising bounds spreads the middle out, but it does so by tying
the top cities together. The cell below prices that trade-off on real indicators rather than
assuming one answer.

Four schemes compared: no winsorising, 1st/99th (current), 5th/95th, and rank-normalisation
(replace each value by its percentile — guarantees an even spread but discards magnitude, so a city
producing ten times another's output is only marginally ahead).


```python
# What do the winsorising choices actually cost?
def _scheme(raw, kind):
    s = np.log1p(raw)
    if kind == "rank":
        return s.rank(pct=True)
    lo_q, hi_q = {"none": (0.0, 1.0), "1/99": (0.01, 0.99), "5/95": (0.05, 0.95)}[kind]
    lo, hi = s.quantile(lo_q), s.quantile(hi_q)
    return (s.clip(lo, hi) - lo) / (hi - lo)

_cities = data_official_Türkiye[data_official_Türkiye["Location_Name"] != "Türkiye"]
_schemes = ["none", "1/99", "5/95", "rank"]

print("Spread achieved, per scheme (main panel, a skewed indicator and a well-behaved one):")
for col in ["agro_greenhouse_prod_ton_perArea", "landuse_core_perArea"]:
    raw = _cities[col].dropna()
    print(f"\n  {col}")
    print(f"    {'scheme':6s} {'%<0.05':>8s} {'%==0':>7s} {'%==1':>7s} {'median':>8s} {'IQR':>7s}")
    for kind in _schemes:
        n = _scheme(raw, kind)
        print(f"    {kind:6s} {(n < 0.05).mean()*100:7.1f}% {(n <= 1e-12).mean()*100:6.1f}% "
              f"{(n >= 1-1e-12).mean()*100:6.1f}% {n.median():8.3f} "
              f"{n.quantile(.75) - n.quantile(.25):7.3f}")

print()
print("And what it costs at the top -- greenhouse output, 2022:")
_g = _cities[_cities["Year"] == 2022][["Location_Name", "agro_greenhouse_prod_ton_perArea"]].dropna()
_raw_all = _cities["agro_greenhouse_prod_ton_perArea"].dropna()
print(f"  {'city':14s}" + "".join(f"{k:>9s}" for k in _schemes))
for city in _g.nlargest(6, "agro_greenhouse_prod_ton_perArea")["Location_Name"]:
    idx = _g[_g["Location_Name"] == city].index[0]
    print(f"  {city:14s}" + "".join(f"{_scheme(_raw_all, k).loc[idx]:9.3f}" for k in _schemes))
print()
print("READ: at 5/95 the top four greenhouse provinces all collapse to exactly 1.000 -- Antalya,")
print("Mersin, Adana and Muğla become indistinguishable in the category where they are the whole")
print("story. Rank-normalisation keeps them apart but flattens magnitude: Antalya produces far more")
print("than Mersin, yet lands only 0.010 above it.")
print()
print("DECISION (2026-09-20): keep 1/99. The floor-bunching in greenhouse is substantively REAL --")
print("most Turkish provinces genuinely have negligible greenhouse agriculture, and an indicator")
print("saying so is telling the truth, not malfunctioning. Consequence to document rather than")
print("engineer away: such an indicator contributes little discrimination to an equal-weighted mean,")
print("so its practical weight is below its nominal share of the category.")
```

## Normalisation — `log1p`, winsorise, pooled min-max

First construction step. Each indicator is transformed independently; raw columns are kept and the
normalised versions are written alongside with a `_norm` suffix, so nothing is destroyed.

**Order: `log1p` → winsorise → min-max.** `log1p` handles skew, winsorising caps whatever extreme
values remain, and min-max maps the result onto [0, 1]. Notes on each step:

- **`log1p`, not `log`** — several indicators contain genuine zeros (a city with no greenhouse in a
  given year), and `log(0)` is undefined. `log1p(x) = log(1+x)` is defined at zero and behaves like
  `log(x)` for large x. Verified: no indicator contains negative values, so this is safe throughout.
  Being monotonic, it never changes a city's rank *within* an indicator — it only changes how much
  weight extreme cities carry when indicators are later averaged together.
- **Applied to every indicator by default.** `LOG1P_SKEW_THRESHOLD = 0.0` transforms all of them;
  raising it to e.g. `1.0` switches to selective transformation of the skewed ones only. Uniform
  treatment avoids an arbitrary cutoff that would need defending, and costs nothing for the few
  near-symmetric indicators. The threshold is a single parameter so the choice stays visible.
- **Thresholds are computed from cities only**, excluding the `Türkiye` aggregate row: the scale
  should be defined by the units actually being compared. Türkiye is then placed on that scale
  (and clipped to it) as a reference line rather than helping to define it.
- **Pooled across all years, not per-year** (Orhan, 2026-09-17). Fixed 2008–2024 thresholds mean a
  score change reflects the city actually changing, not a shift relative to whoever else was
  measured that year — which is what the DiD needs. GFSI uses the same fixed-threshold approach for
  exactly this reason.
- **Cost-direction flips are NOT applied here.** Water, waste, energy and land-use-fallow are
  cost-framed, but flipping happens at aggregation as `1 − x`, so these normalised columns stay
  directly comparable and the flip remains visible and reversible.

Both denominator tracks are normalised, but they are never mixed: `_perArea` and `_perHousehold`
columns are normalised separately and feed separate indices (see Diagnostic 6 for why averaging
them would produce a measure contaminated in both dimensions).


```python
LOG1P_SKEW_THRESHOLD = 0.0        # 0.0 -> transform every indicator; raise (e.g. 1.0) to be selective
WINSOR_LIMITS = (0.01, 0.99)      # cap at 1st / 99th percentile before scaling

# indicator_columns() is defined in the pre-normalisation profile cell above.

def normalise_indicators(df, panel_name):
    """log1p -> winsorise -> pooled min-max. Returns (df with _norm columns, audit table)."""
    out = df.copy()
    is_city = df["Location_Name"] != "Türkiye"
    audit = []

    for col in indicator_columns(df):
        raw = df[col]
        skew_before = raw[is_city].skew()

        transformed = skew_before > LOG1P_SKEW_THRESHOLD
        s = np.log1p(raw) if transformed else raw.copy()

        lo = s[is_city].quantile(WINSOR_LIMITS[0])
        hi = s[is_city].quantile(WINSOR_LIMITS[1])
        n_clipped = int(((s < lo) | (s > hi)).sum())

        s_clipped = s.clip(lo, hi)
        out[col + "_norm"] = (s_clipped - lo) / (hi - lo) if hi > lo else np.nan

        audit.append({
            "panel": panel_name, "indicator": col, "log1p": transformed,
            "skew_before": round(skew_before, 2),
            "skew_after": round(s[is_city].skew(), 2),
            "clipped": n_clipped,
        })
    return out, pd.DataFrame(audit)

data_official_Türkiye, audit_main = normalise_indicators(data_official_Türkiye, "main")
data_official_Türkiye_extended, audit_ext = normalise_indicators(data_official_Türkiye_extended, "extended")
normalisation_audit = pd.concat([audit_main, audit_ext], ignore_index=True)

print(f"Normalised {len(normalisation_audit)} indicators "
      f"({len(audit_main)} main, {len(audit_ext)} extended).")
print(f"log1p applied to {int(normalisation_audit['log1p'].sum())} of {len(normalisation_audit)}; "
      f"threshold = {LOG1P_SKEW_THRESHOLD}, winsorised at {WINSOR_LIMITS}.")
print()
print("Skew reduction achieved by log1p (cities only):")
print(f"  mean |skew| before : {normalisation_audit['skew_before'].abs().mean():.2f}")
print(f"  mean |skew| after  : {normalisation_audit['skew_after'].abs().mean():.2f}")
print()
normalisation_audit.sort_values("skew_before", ascending=False)
```

```python
# Verification: every _norm column must sit in [0, 1], and the transformation must not have
# changed any city's rank within an indicator (log1p and min-max are both monotonic).
from scipy.stats import spearmanr

problems = []
for panel, df in [("main", data_official_Türkiye), ("extended", data_official_Türkiye_extended)]:
    for col in indicator_columns(df):
        n = df[col + "_norm"]
        valid = n.dropna()
        if len(valid) and (valid.min() < -1e-9 or valid.max() > 1 + 1e-9):
            problems.append(f"{panel}/{col}: out of range [{valid.min():.4f}, {valid.max():.4f}]")
        # rank preservation, checked on unclipped rows only (winsorising deliberately ties the tails)
        pair = pd.DataFrame({"raw": df[col], "norm": n}).dropna()
        interior = pair[(pair["norm"] > 0) & (pair["norm"] < 1)]
        if len(interior) > 10:
            rho = spearmanr(interior["raw"], interior["norm"]).statistic
            if rho < 0.9999:
                problems.append(f"{panel}/{col}: rank not preserved (rho={rho:.6f})")
        if n.isna().sum() != df[col].isna().sum():
            problems.append(f"{panel}/{col}: NaN count changed")

print("Range check, rank preservation and NaN preservation:")
print("  PASS - nothing to report" if not problems else "\n".join("  FAIL " + p for p in problems))
print()
norm_cols = [c for c in data_official_Türkiye.columns if c.endswith("_norm")]
print(f"main: {len(norm_cols)} normalised columns, shape {data_official_Türkiye.shape}")
print(data_official_Türkiye[norm_cols].describe().loc[["min", "50%", "max"]].T.round(3).to_string())
```

## Aggregation — cost-direction flips and category structure

Full reasoning for every decision this section implements is in
`agent_note_econometrics_FSOI.md` (Part 1: "Category structure", "Two-track index
structure") — summarised here only enough to make the code below legible.

**Five categories, not six** (water and waste merged into one, "municipal burden",
2026-09-22): `production`, `land_use`, `municipal_burden` (water + waste), `external_input`
(fertiliser + agricultural electricity), `market`.

**Cost-direction flips** (`score = 1 − x` on the `_norm` column; benefit indicators keep
`score = x`): both municipal-burden indicators, both external-input indicators, and
`landuse_fallow_km2` alone within land-use. Applied here rather than at normalisation, so
the `_norm` columns stay directly comparable and the flip stays visible and reversible.

**Two tracks, one pipeline:**
- **Track C** — all five categories, 2008–2020, 14 indicator pairs. The FSOI as defined —
  the descriptive core: levels, city rankings, distribution.
- **Track A** — four categories (production, land_use, municipal_burden *waste-only*,
  external_input *fertiliser-only*), 2008–2024, 9 indicator pairs. The DiD-ready track.

Both are built from `_perHousehold` columns only (the headline denominator). `_perArea` is
mirrored afterward as the robustness track and never mixed with `_perHousehold` in one
aggregation (Diagnostic 6: the two tracks are contaminated in opposite dimensions).

```python
# Cost-direction flips (2026-09-21/22 decisions; see agent note "Category structure").
# score = 1 - norm for cost/burden indicators ("lower raw value is better"), score = norm
# for benefit indicators. Every _norm column gets a _score twin, so aggregation only ever
# has to average "_score" columns regardless of direction -- the flip is applied once, here.

COST_INDICATORS = {
    "wasteCollected_1000ton",       # municipal burden
    "water_drainage",               # municipal burden
    "fertilizer_use",               # external input
    "electricity_agriculture_mwh",  # external input
    "landuse_fallow_km2",           # land-use -- the one cost-framed member
}

def apply_cost_flips(df):
    """For every *_norm column, add a *_score column: 1 - norm if its base indicator is in
    COST_INDICATORS, else norm unchanged. Never mutates the _norm column itself."""
    out = df.copy()
    flipped, kept = [], []
    for col in df.columns:
        if not col.endswith("_norm"):
            continue
        base = col[:-len("_norm")]
        indicator = None
        for suffix in ("_perArea", "_perHousehold"):
            if base.endswith(suffix):
                indicator = base[: -len(suffix)]
                break
        assert indicator is not None, f"normalised column with unexpected name: {col}"
        if indicator in COST_INDICATORS:
            out[base + "_score"] = 1 - out[col]
            flipped.append(indicator)
        else:
            out[base + "_score"] = out[col]
            kept.append(indicator)
    return out, sorted(set(flipped)), sorted(set(kept))

data_official_Türkiye, flipped_main, kept_main = apply_cost_flips(data_official_Türkiye)
data_official_Türkiye_extended, flipped_ext, kept_ext = apply_cost_flips(data_official_Türkiye_extended)

print(f"Main panel     -- flipped (cost): {flipped_main}")
print(f"               -- kept (benefit): {kept_main}")
print(f"Extended panel -- flipped (cost): {flipped_ext}")
print(f"               -- kept (benefit): {kept_ext}")

# The flip list must match the settled decision exactly -- nothing more, nothing less.
all_flipped = set(flipped_main) | set(flipped_ext)
assert all_flipped == COST_INDICATORS, f"flip mismatch -- expected {COST_INDICATORS}, got {all_flipped}"
print("\nFlip list matches the settled decision exactly (waste, water, fertiliser, "
      "electricity, land-use fallow -- nothing else).")
```

### Category definitions

```python
# Category -> member indicator base names (no perArea/perHousehold suffix, no _score
# suffix). A category can draw from both panels (municipal_burden: waste is main, water is
# extended) -- that asymmetry is exactly why Track C needs a merge and Track A does not.

MAIN_INDICATORS = [
    "agro_greenhouse_prod_ton", "total_agro_production_ton",
    "landuse_core", "landuse_fallow_km2", "landuse_greenhouse_km2",
    "landuse_longtermCrops_km2", "landuse_vegetables_km2",
    "wasteCollected_1000ton", "fertilizer_use",
]
EXTENDED_INDICATORS = [
    "agro_crop_1000USD", "agro_livestock_1000USD", "agro_animalproducts_1000USD",
    "water_drainage", "electricity_agriculture_mwh",
]
assert len(MAIN_INDICATORS) == 9 and len(EXTENDED_INDICATORS) == 5, \
    "expected 9 main + 5 extended = 14 indicator pairs total"

# Track C: five categories, every indicator.
CATEGORY_MAP_C = {
    "production":       ["agro_greenhouse_prod_ton", "total_agro_production_ton"],
    "land_use":         ["landuse_core", "landuse_fallow_km2", "landuse_greenhouse_km2",
                          "landuse_longtermCrops_km2", "landuse_vegetables_km2"],
    "municipal_burden": ["wasteCollected_1000ton", "water_drainage"],
    "external_input":   ["fertilizer_use", "electricity_agriculture_mwh"],
    "market":           ["agro_crop_1000USD", "agro_livestock_1000USD", "agro_animalproducts_1000USD"],
}

# Track A: four categories, main-panel indicators only (2026-09-22 resolution, option (a) --
# each category uses everything it has in the main panel; market has no main-panel member,
# so it drops out entirely rather than being represented partially).
CATEGORY_MAP_A = {
    "production":       ["agro_greenhouse_prod_ton", "total_agro_production_ton"],
    "land_use":         ["landuse_core", "landuse_fallow_km2", "landuse_greenhouse_km2",
                          "landuse_longtermCrops_km2", "landuse_vegetables_km2"],
    "municipal_burden": ["wasteCollected_1000ton"],  # waste only -- water is extended-panel
    "external_input":   ["fertilizer_use"],          # fertiliser only -- electricity is extended-panel
}

# Every indicator named in either map must be one we actually built.
_all_known = set(MAIN_INDICATORS) | set(EXTENDED_INDICATORS)
for _name, _cmap in [("CATEGORY_MAP_C", CATEGORY_MAP_C), ("CATEGORY_MAP_A", CATEGORY_MAP_A)]:
    _used = {ind for members in _cmap.values() for ind in members}
    _unknown = _used - _all_known
    assert not _unknown, f"{_name} references unknown indicators: {_unknown}"

print(f"CATEGORY_MAP_C: {len(CATEGORY_MAP_C)} categories, "
      f"{sum(len(v) for v in CATEGORY_MAP_C.values())} indicator pairs -- expect 5, 14")
print(f"CATEGORY_MAP_A: {len(CATEGORY_MAP_A)} categories, "
      f"{sum(len(v) for v in CATEGORY_MAP_A.values())} indicator pairs -- expect 4, 9")
```

### Building the composite: category sub-indices → equal-weighted FSOI

Same function for both tracks: a category sub-index is the simple mean of its member
`_score` columns, and the composite is the simple mean of the category sub-indices — equal
weight at both levels, per the settled aggregation method (equal-weighted sum, not
TOPSIS).

```python
def build_fsoi(df, category_map, suffix, year_range=None):
    """Category sub-indices (mean of member _score columns) and the equal-weighted composite.

    suffix: "_perHousehold" (headline) or "_perArea" (robustness track) -- never mixed.
    year_range: optional inclusive (min_year, max_year) filter, applied before scoring.
    Returns Year, Location_Name, Treated, Treated_Label, one column per category, and FSOI
    (the unweighted mean across categories -- equal category weight).
    """
    if year_range is not None:
        lo, hi = year_range
        df = df[df["Year"].between(lo, hi)]

    out = df[["Year", "Location_Name", "Treated", "Treated_Label"]].copy()
    for category, indicators in category_map.items():
        cols = [f"{ind}{suffix}_score" for ind in indicators]
        missing = [c for c in cols if c not in df.columns]
        assert not missing, f"category '{category}' missing columns: {missing}"
        out[category] = df[cols].mean(axis=1)

    category_cols = list(category_map.keys())
    out["FSOI"] = out[category_cols].mean(axis=1)
    return out
```

### Track C — the index: five categories, 2008–2020

```python
# Track C needs both panels together (market and water are extended-panel). Merge on all
# four columns confirmed identical across both dataframes by design (agent note, "Shared
# columns") rather than just Year+Location_Name, so a Treated/Treated_Label mismatch between
# panels breaks the merge loudly instead of silently duplicating columns.
panel_C = data_official_Türkiye.merge(
    data_official_Türkiye_extended,
    on=["Year", "Location_Name", "Treated", "Treated_Label"],
    how="inner", validate="one_to_one",
)
print(f"Merged panel for Track C: {panel_C.shape[0]} rows "
      f"(main {data_official_Türkiye.shape[0]}, extended {data_official_Türkiye_extended.shape[0]})")

FSOI_C = build_fsoi(panel_C, CATEGORY_MAP_C, suffix="_perHousehold", year_range=(2008, 2020))

print(f"\nTrack C (FSOI_C): {FSOI_C.shape[0]} rows, years {sorted(FSOI_C['Year'].unique())}")
assert FSOI_C["FSOI"].between(0, 1, inclusive="both").all(), "FSOI_C out of [0, 1] range"
assert set(FSOI_C["Year"].unique()) == {2008, 2010, 2012, 2014, 2016, 2018, 2020}, \
    "Track C should be exactly the 7 biennial years 2008-2020"
FSOI_C.describe().round(3)
```

### Do Track C and Track A agree, over the years both exist? (convergent-validity check)

Both tracks share 2008–2020. If Track A's four-category composite ranks cities the way
Track C's full five-category composite does *in that shared window*, that is evidence the
four-category version is a reasonable stand-in for the defined index — which is what would
license reading Track A's 2022/2024 values as an extension of the same construct rather
than an unrelated four-category measure. If they diverge, that licence doesn't hold, and A's
later years should be reported on their own terms rather than as "the FSOI."

```python
from scipy.stats import spearmanr

overlap_years = sorted(set(FSOI_C["Year"]) & set(FSOI_A["Year"]))
rows = []
for yr in overlap_years:
    c_yr = FSOI_C[(FSOI_C["Year"] == yr) & (FSOI_C["Location_Name"] != "Türkiye")]
    a_yr = FSOI_A[(FSOI_A["Year"] == yr) & (FSOI_A["Location_Name"] != "Türkiye")]
    merged_yr = c_yr[["Location_Name", "FSOI"]].merge(
        a_yr[["Location_Name", "FSOI"]], on="Location_Name", suffixes=("_C", "_A"))
    rho = spearmanr(merged_yr["FSOI_C"], merged_yr["FSOI_A"]).statistic
    rows.append({"Year": yr, "n_cities": len(merged_yr), "spearman_rho": round(rho, 3)})

convergence = pd.DataFrame(rows)
print("Track C vs Track A rank agreement, shared years (2008-2020):")
print(convergence.to_string(index=False))
print(f"\nMean rho across shared years: {convergence['spearman_rho'].mean():.3f}")

# Which cities move the most in rank between the two tracks, in the most recent shared year?
yr = overlap_years[-1]
c_yr = FSOI_C[(FSOI_C["Year"] == yr) & (FSOI_C["Location_Name"] != "Türkiye")].copy()
a_yr = FSOI_A[(FSOI_A["Year"] == yr) & (FSOI_A["Location_Name"] != "Türkiye")].copy()
c_yr["rank_C"] = c_yr["FSOI"].rank(ascending=False)
a_yr["rank_A"] = a_yr["FSOI"].rank(ascending=False)
rank_compare = c_yr[["Location_Name", "Treated_Label", "rank_C"]].merge(
    a_yr[["Location_Name", "rank_A"]], on="Location_Name")
rank_compare["rank_shift"] = (rank_compare["rank_C"] - rank_compare["rank_A"]).abs()

print(f"\nBiggest rank movers, {yr} (Track C five-category rank vs Track A four-category rank):")
print(rank_compare.nlargest(10, "rank_shift").to_string(index=False))
```

## Difference-in-Differences — did Law 6360 change food sovereignty?

Everything above (flips, category sub-indices, the composite, LOCO, the convergence check)
was groundwork. This is the causal question the whole pipeline exists to answer.

**Track A is the DiD track** (2008–2024, four categories, six post-treatment periods).
Track C's post-treatment years are 2014/2016/2018/2020 and 2020 is COVID, leaving
effectively three clean post-treatment points — not enough for a credible event study (see
"Two-track index structure" in the agent note).

**Primary control group: non-metropolitan only.** Track A's composite includes
`municipal_burden` (waste), a municipal-service variable, and the settled rule (agent note,
"Control groups: which is clean, and for which variables") is that old-metropolitan is not a
clean control for municipal-service variables — Law 6360 extended metropolitan boundaries
for *existing* metros too (implied coverage +8.9pp at the reform), so a treated-vs-old-metro
comparison on a composite containing municipal_burden would understate whatever effect
exists. Old-metropolitan appears below only as a labelled, contaminated robustness
comparison, never the primary estimate.

**Specification:** two-way fixed effects via OLS — `FSOI ~ Treated×Post + City FE + Year
FE` — with the main `Treated` and `Post` effects omitted deliberately, since they are exactly
collinear with the city and year fixed effects respectively (treatment status never varies
within a city, and `Post` never varies within a year) and would either be dropped or make
the design matrix rank-deficient if included alongside the FE. Standard errors are
cluster-robust by city. `Post = 1` for Year ≥ 2014 (the reform's effective date via the
March 2014 local elections — 2012 stays pre-treatment, matching this pipeline's convention
throughout).

**Caveat carried from the panel design itself, not fixable here:** only three pre-treatment
points exist in this data at all (2008, 2010, 2012), in any track. The event-study check
below tests what it can with that, but a short pre-period is a limitation of the underlying
TÜİK biennial panel, not of this specification — treat the pre-trend result as suggestive,
not decisive.

**Note:** `statsmodels` is used here (installed in this environment, 0.14.5) and is not
currently listed in `requirements.txt` — flagged to `thesis_log_main_agent` to add, since
that file is repo-root and outside this strand's scope.

```python
import statsmodels.formula.api as smf

# Two-group DiD sample: non-metropolitan (control) vs new-metropolitan (treated) only --
# old-metropolitan excluded from the primary specification per the settled control-group rule.
did_sample = FSOI_A[FSOI_A["Treated_Label"].isin(["Non-metropolitan", "New-metropolitan (2012)"])].copy()
did_sample["Treated_bin"] = (did_sample["Treated_Label"] == "New-metropolitan (2012)").astype(int)
did_sample["Post"] = (did_sample["Year"] >= 2014).astype(int)
did_sample["TreatedXPost"] = did_sample["Treated_bin"] * did_sample["Post"]

n_treated = did_sample.loc[did_sample["Treated_bin"] == 1, "Location_Name"].nunique()
n_control = did_sample.loc[did_sample["Treated_bin"] == 0, "Location_Name"].nunique()
print(f"DiD sample: {len(did_sample)} obs, {n_treated + n_control} cities "
      f"({n_treated} treated / new-metropolitan, {n_control} control / non-metropolitan), "
      f"{did_sample['Year'].nunique()} years ({sorted(did_sample['Year'].unique())})")

# Treated_bin and Post are deliberately NOT included as separate main-effect terms -- they
# are exactly collinear with C(Location_Name) and C(Year) respectively (treatment status is
# time-invariant per city; Post is city-invariant per year), so the standard two-way-FE DiD
# formula includes only the interaction on top of both fixed-effect sets.
did_model = smf.ols("FSOI ~ TreatedXPost + C(Location_Name) + C(Year)", data=did_sample).fit(
    cov_type="cluster", cov_kwds={"groups": did_sample["Location_Name"]})

did_coef = did_model.params["TreatedXPost"]
did_se = did_model.bse["TreatedXPost"]
did_p = did_model.pvalues["TreatedXPost"]
print(f"\nPRIMARY DiD ESTIMATE (TreatedXPost): {did_coef:+.4f} (SE {did_se:.4f}, "
      f"p = {did_p:.4f}, n = {int(did_model.nobs)}, R\u00b2 = {did_model.rsquared:.3f})")
print(f"95% CI: [{did_coef - 1.96*did_se:+.4f}, {did_coef + 1.96*did_se:+.4f}]")
print(f"\nCaution on inference: {n_treated} treated clusters is on the low side for "
      f"asymptotic cluster-robust standard errors (rule-of-thumb minimum is often put at "
      f"30-50 clusters, and specifically few TREATED clusters is its own known issue -- "
      f"Bertrand/Duflo/Mullainathan 2004). A wild cluster bootstrap would be the standard "
      f"fix; not implemented here -- flagged as a robustness item, not run.")
```

### Wild cluster bootstrap — correcting inference for 14 treated clusters

The primary estimate's cluster-robust p-value (0.0031) relies on asymptotics in the number
of clusters, and 14 treated clusters is squarely the regime where that asymptotic
approximation is known to be unreliable (Cameron, Gelbach & Miller 2008; Bertrand, Duflo &
Mullainathan 2004) — flagged as a blocking item when the primary regression was first run,
not a nice-to-have. This runs the WCR (wild cluster bootstrap, restricted) procedure:
Rademacher weights (±1, drawn per city) applied to the residuals of the *null-restricted*
model (no `TreatedXPost` term), refit under the unrestricted specification for each draw,
and the observed t-statistic compared against the resulting bootstrap distribution.

```python
import numpy as np
import patsy
from statsmodels.api import OLS

np.random.seed(20260922)  # dated, documented -- reproducible, not tuned to a result

y_mat, X_mat = patsy.dmatrices("FSOI ~ TreatedXPost + C(Location_Name) + C(Year)",
                                data=did_sample, return_type="dataframe")
treated_col = "TreatedXPost"
X_restricted = X_mat.drop(columns=[treated_col])

restricted_fit = OLS(y_mat, X_restricted).fit()
y_hat_restricted = restricted_fit.fittedvalues.values.ravel()
resid_restricted = restricted_fit.resid.values.ravel()

clusters = did_sample["Location_Name"].values
unique_clusters = np.unique(clusters)
n_boot = 999

t_obs = did_coef / did_se
boot_t = np.empty(n_boot)

for b in range(n_boot):
    cluster_weights = dict(zip(unique_clusters, np.random.choice([-1, 1], size=len(unique_clusters))))
    w = np.array([cluster_weights[c] for c in clusters])
    y_boot = y_hat_restricted + resid_restricted * w
    boot_fit = OLS(y_boot, X_mat).fit(cov_type="cluster", cov_kwds={"groups": clusters})
    boot_t[b] = boot_fit.params[treated_col] / boot_fit.bse[treated_col]

wild_p = np.mean(np.abs(boot_t) >= np.abs(t_obs))
n_treated_cities = did_sample.loc[did_sample["Treated_bin"] == 1, "Location_Name"].nunique()

print(f"Wild cluster bootstrap (Rademacher, WCR, B={n_boot}, {len(unique_clusters)} clusters "
      f"[{n_treated_cities} treated]):")
print(f"  observed t-statistic: {t_obs:.3f}")
print(f"  bootstrap p-value:    {wild_p:.4f}")
print(f"  asymptotic cluster-robust p-value (for comparison): {did_p:.4f}")
print()
if wild_p < 0.05:
    print("Bootstrap CONFIRMS significance at p<0.05.")
else:
    print("Bootstrap DOES NOT confirm significance at p<0.05 -- the asymptotic p-value overstated it.")
```

### Pre-trend check — event study relative to 2012

Year-specific treatment effects relative to 2012 (the last pre-treatment year), holding city
and year fixed effects. Tests whether treated and control cities were already diverging
before the reform (2008, 2010 vs. 2012) and traces the effect year by year afterward. With
only two pre-treatment comparison points, this is a weak test of parallel trends, not a
strong one — say so, don't oversell it either way.

```python
event_years = sorted(did_sample["Year"].unique())
ref_year = 2012

for yr in event_years:
    if yr == ref_year:
        continue
    did_sample[f"TxY_{yr}"] = ((did_sample["Year"] == yr) & (did_sample["Treated_bin"] == 1)).astype(int)

interaction_terms = [f"TxY_{yr}" for yr in event_years if yr != ref_year]
event_formula = "FSOI ~ " + " + ".join(interaction_terms) + " + C(Location_Name) + C(Year)"
event_model = smf.ols(event_formula, data=did_sample).fit(
    cov_type="cluster", cov_kwds={"groups": did_sample["Location_Name"]})

event_rows = []
for yr in event_years:
    if yr == ref_year:
        event_rows.append({"Year": yr, "coef": 0.0, "se": 0.0, "p": np.nan, "period": "ref (2012)"})
        continue
    term = f"TxY_{yr}"
    event_rows.append({
        "Year": yr, "coef": event_model.params[term], "se": event_model.bse[term],
        "p": event_model.pvalues[term], "period": "pre" if yr < 2014 else "post",
    })
event_study = pd.DataFrame(event_rows)
print("Event-study: treatment effect relative to 2012, by year:")
print(event_study.round(4).to_string(index=False))

pre_period = event_study[event_study["period"] == "pre"]
pre_clear = (pre_period["p"] > 0.05).all()
print(f"\nPre-trend check ({len(pre_period)} pre-treatment comparison point(s) besides the "
      f"reference year): "
      + ("no pre-period coefficient is significant at p<0.05 -- consistent with parallel "
         "pre-trends, though only one comparison point (2008 vs 2012) makes this a weak test."
         if pre_clear else
         "WARNING -- a pre-period coefficient is significant at p<0.05. Parallel trends is "
         "questionable and the DiD estimate above should be read with that caveat attached."))
```

### Which municipal_burden confound? Coverage expansion vs. the tariff waiver have different time signatures

Both documented confounds land on `municipal_burden`, but they don't predict the same shape
over time. **Coverage-boundary expansion** happened at the reform itself — metropolitan
status extended service boundaries province-wide starting 2014 — so it predicts a **level
shift right at 2014**, roughly flat afterward. **The tariff waiver** ran 2014–2019 and then
ended, so if capped costs suppressed the cost-framed indicators during the waiver and
uncapped costs raised them afterward, it predicts an effect that is **small or absent through
2018 and grows from 2020/2022 onward** — the opposite time shape. Re-runs the event-study
design already built above, with `municipal_burden` as the outcome instead of `FSOI`, reusing
the same `TxY_*` treatment-year interaction dummies.

```python
mb_event_formula = "municipal_burden ~ " + " + ".join(interaction_terms) + " + C(Location_Name) + C(Year)"
mb_event_model = smf.ols(mb_event_formula, data=did_sample).fit(
    cov_type="cluster", cov_kwds={"groups": did_sample["Location_Name"]})

mb_event_rows = []
for yr in event_years:
    if yr == ref_year:
        mb_event_rows.append({"Year": yr, "coef": 0.0, "se": 0.0, "p": np.nan, "period": "ref (2012)"})
        continue
    term = f"TxY_{yr}"
    mb_event_rows.append({
        "Year": yr, "coef": mb_event_model.params[term], "se": mb_event_model.bse[term],
        "p": mb_event_model.pvalues[term], "period": "pre" if yr < 2014 else "post",
    })
mb_event_study = pd.DataFrame(mb_event_rows)
print("Treatment effect on municipal_burden ALONE, relative to 2012, by year:")
print(mb_event_study.round(4).to_string(index=False))

# Onset timing (2014) does NOT discriminate between the two mechanisms: a permanent boundary
# redefinition and a price cap taking effect both predict an immediate 2014 jump. What
# distinguishes them is the FADE -- a boundary redefinition has no built-in reason to fade;
# a price cap that lifts in 2019/2020 does (see the markdown correction below this cell).
pre2020_post = mb_event_study[mb_event_study["Year"].isin([2014, 2016, 2018])]["coef"]
post2020 = mb_event_study[mb_event_study["Year"].isin([2020, 2022, 2024])]["coef"]
fade_pct = 100 * (1 - post2020.mean() / pre2020_post.mean()) if pre2020_post.mean() != 0 else float("nan")
print(f"\nMean effect, 2014-2018 (waiver active): {pre2020_post.mean():+.4f}")
print(f"Mean effect, 2020-2024 (waiver ended):   {post2020.mean():+.4f}")
print(f"Fade from 2014-2018 to 2020-2024: {fade_pct:.0f}%")
print("\nSee the markdown cell immediately below for the corrected reading of this shape --")
print("the 2014 onset does not distinguish the two mechanisms; the fade does, and points")
print("toward a price-driven waiver effect rather than a permanent boundary-expansion effect")
print("-- confounded with COVID landing at the same date.")
```

production, land-use and external-input show **no detectable treatment effect** in this
design.

## Robustness: equal-weight-per-indicator (flat) aggregation

**Task from Orhan, via main agent, 2026-09-23.** The primary specification weights
categories equally, then indicators equally within a category — so an indicator's overall
weight is `1/(N_categories × k_category)`. In Track A, `municipal_burden` and
`external_input` each collapsed to exactly **one** indicator once extended-panel indicators
dropped out (waste alone, fertiliser alone), so each of those single indicators carries
**25%** of the entire composite on its own (`1/4 categories × 1/1`), while each of
`land_use`'s five indicators carries only **5%** (`1/4 × 1/5`) — a 5× disparity. That is a
plausible driver of `municipal_burden` dominating the DiD, independent of any confound
story: it may simply be the most heavily-weighted single indicator in the composite by
construction.

**This section builds the flat alternative — every indicator gets equal weight, 1/9 in
Track A, 1/14 in Track C — and reruns the composite, LOCO, and primary DiD under it.** The
category-weighted results above remain the primary specification and are not replaced;
this is a side-by-side comparison, reported as such in the note and eventual write-up.

```python
def build_fsoi_flat(df, category_map, suffix, year_range=None):
    """Equal-weight-per-INDICATOR alternative to build_fsoi. Every member indicator across
    the flattened category_map gets identical weight 1/M (M = total indicator count), rather
    than 1/(N_categories x k_category). Category sub-indices are still computed and reported
    (informational, same as build_fsoi) but do NOT feed the composite -- FSOI here is the
    flat mean of every indicator''s own _score column directly."""
    if year_range is not None:
        lo, hi = year_range
        df = df[df["Year"].between(lo, hi)]

    out = df[["Year", "Location_Name", "Treated", "Treated_Label"]].copy()
    all_indicators = [ind for members in category_map.values() for ind in members]
    for category, indicators in category_map.items():
        cols = [f"{ind}{suffix}_score" for ind in indicators]
        out[category] = df[cols].mean(axis=1)  # informational only -- not used for FSOI below

    flat_cols = [f"{ind}{suffix}_score" for ind in all_indicators]
    out["FSOI"] = df[flat_cols].mean(axis=1)
    return out


# Per-indicator weight table, both schemes, both tracks -- the disparity claimed above,
# computed rather than asserted.
weight_rows = []
for track_name, cmap, n_total in [("Track A", CATEGORY_MAP_A, 9), ("Track C", CATEGORY_MAP_C, 14)]:
    for category, indicators in cmap.items():
        cat_weight_share = 1 / len(cmap) / len(indicators)
        flat_weight_share = 1 / n_total
        for ind in indicators:
            weight_rows.append({
                "track": track_name, "category": category, "indicator": ind,
                "category_weighted_%": round(100 * cat_weight_share, 1),
                "flat_weighted_%": round(100 * flat_weight_share, 1),
            })
weight_table = pd.DataFrame(weight_rows)
print("Per-indicator weight under each scheme:")
print(weight_table.to_string(index=False))
print(f"\nTrack A category-weighted range: {weight_table[weight_table.track=='Track A']['category_weighted_%'].min()}%"
      f" to {weight_table[weight_table.track=='Track A']['category_weighted_%'].max()}% "
      f"(ratio {weight_table[weight_table.track=='Track A']['category_weighted_%'].max() / weight_table[weight_table.track=='Track A']['category_weighted_%'].min():.1f}x)")
```

### Flat-weighted composites and group means

### 1. TOPSIS vs. equal-weighted sum

Yilmaz (2025)'s Entropy-TOPSIS precedent: vector-normalise each indicator (`x / sqrt(sum
x^2)`, distinct from the min-max already applied during normalisation), apply weights, then
rank by closeness to an ideal solution rather than by a simple average. To isolate the
**aggregation-method** comparison specifically (not re-litigate indicator selection or
weights), this uses the exact same per-indicator weights as the primary equal-weighted-sum
specification (the category-then-within weights already computed in the flat-weighting
section above) — only the aggregation formula changes. Reference points (ideal best/worst)
are computed from the full city panel (cities only, all years), matching this pipeline's
pooled-threshold philosophy elsewhere, so they don't shift within the comparison.

```python
def topsis_composite(df, category_map, suffix, year_range=None):
    """TOPSIS closeness-to-ideal composite, using the SAME per-indicator weights as the
    primary equal-weighted-sum spec (category-then-within), so this isolates the
    aggregation-method choice rather than also changing weights. Ideal best/worst points are
    computed from the full city panel (cities only), fixed before any year_range filter."""
    all_indicators = [ind for members in category_map.values() for ind in members]
    score_cols = [f"{ind}{suffix}_score" for ind in all_indicators]

    weights = {}
    for category, members in category_map.items():
        w = 1 / len(category_map) / len(members)
        for ind in members:
            weights[f"{ind}{suffix}_score"] = w
    w_vec = np.array([weights[c] for c in score_cols])

    cities_only = df[df["Location_Name"] != "Türkiye"]
    norms = np.sqrt((cities_only[score_cols] ** 2).sum(axis=0))
    V_cities = (cities_only[score_cols] / norms) * w_vec
    ideal_best = V_cities.max(axis=0)
    ideal_worst = V_cities.min(axis=0)

    if year_range is not None:
        lo, hi = year_range
        df = df[df["Year"].between(lo, hi)]

    out = df[["Year", "Location_Name", "Treated", "Treated_Label"]].copy()
    V = (df[score_cols] / norms) * w_vec
    s_best = np.sqrt(((V - ideal_best) ** 2).sum(axis=1))
    s_worst = np.sqrt(((V - ideal_worst) ** 2).sum(axis=1))
    out["FSOI"] = s_worst / (s_best + s_worst)
    return out


TOPSIS_A = topsis_composite(data_official_Türkiye, CATEGORY_MAP_A, suffix="_perHousehold", year_range=None)
assert TOPSIS_A["FSOI"].between(0, 1, inclusive="both").all(), "TOPSIS_A out of [0, 1]"

topsis_groups = TOPSIS_A[TOPSIS_A["Location_Name"] != "Türkiye"].groupby("Treated_Label", observed=True)["FSOI"].mean().round(3)
sum_groups = FSOI_A[FSOI_A["Location_Name"] != "Türkiye"].groupby("Treated_Label", observed=True)["FSOI"].mean().round(3)
print("Group means, TOPSIS vs equal-weighted-sum (Track A, pooled all years):")
print(pd.concat([sum_groups.rename("equal_weighted_sum"), topsis_groups.rename("topsis")], axis=1))

rank_corr_topsis = spearmanr(
    FSOI_A[FSOI_A["Location_Name"] != "Türkiye"].groupby("Location_Name")["FSOI"].mean(),
    TOPSIS_A[TOPSIS_A["Location_Name"] != "Türkiye"].groupby("Location_Name")["FSOI"].mean(),
).statistic
print(f"\nCity-level rank correlation, equal-weighted-sum vs TOPSIS: rho = {rank_corr_topsis:.3f}")
```

```python
# DiD under TOPSIS, primary and LOCO (same non-metro-only control sample structure).
did_sample_topsis = TOPSIS_A[TOPSIS_A["Treated_Label"].isin(
    ["Non-metropolitan", "New-metropolitan (2012)"])].copy()
did_sample_topsis["Treated_bin"] = (did_sample_topsis["Treated_Label"] == "New-metropolitan (2012)").astype(int)
did_sample_topsis["Post"] = (did_sample_topsis["Year"] >= 2014).astype(int)
did_sample_topsis["TreatedXPost"] = did_sample_topsis["Treated_bin"] * did_sample_topsis["Post"]

did_model_topsis = smf.ols("FSOI ~ TreatedXPost + C(Location_Name) + C(Year)", data=did_sample_topsis).fit(
    cov_type="cluster", cov_kwds={"groups": did_sample_topsis["Location_Name"]})
coef_topsis, se_topsis, p_topsis = (did_model_topsis.params["TreatedXPost"], did_model_topsis.bse["TreatedXPost"],
                                     did_model_topsis.pvalues["TreatedXPost"])
print(f"PRIMARY DiD, TOPSIS: {coef_topsis:+.4f} (SE {se_topsis:.4f}, p = {p_topsis:.4f}, n = {int(did_model_topsis.nobs)})")
print(f"PRIMARY DiD, equal-weighted-sum (for comparison): {did_coef:+.4f} (SE {did_se:.4f}, p = {did_p:.4f})")

TOPSIS_A_loco_mb = topsis_composite(data_official_Türkiye, CATEGORY_MAP_A_LOCO_MB, suffix="_perHousehold", year_range=None)
did_sample_topsis_loco = TOPSIS_A_loco_mb[TOPSIS_A_loco_mb["Treated_Label"].isin(
    ["Non-metropolitan", "New-metropolitan (2012)"])].copy()
did_sample_topsis_loco["Treated_bin"] = (did_sample_topsis_loco["Treated_Label"] == "New-metropolitan (2012)").astype(int)
did_sample_topsis_loco["Post"] = (did_sample_topsis_loco["Year"] >= 2014).astype(int)
did_sample_topsis_loco["TreatedXPost"] = did_sample_topsis_loco["Treated_bin"] * did_sample_topsis_loco["Post"]

did_model_topsis_loco = smf.ols("FSOI ~ TreatedXPost + C(Location_Name) + C(Year)", data=did_sample_topsis_loco).fit(
    cov_type="cluster", cov_kwds={"groups": did_sample_topsis_loco["Location_Name"]})
coef_topsis_loco = did_model_topsis_loco.params["TreatedXPost"]
p_topsis_loco = did_model_topsis_loco.pvalues["TreatedXPost"]
print(f"\nTOPSIS DiD WITHOUT municipal_burden: {coef_topsis_loco:+.4f} (p = {p_topsis_loco:.4f})")
print(f"TOPSIS DiD WITH municipal_burden:    {coef_topsis:+.4f} (p = {p_topsis:.4f})")
print(f"Sign check: {'same direction, survives' if (coef_topsis_loco < 0) == (coef_topsis < 0) else 'SIGN FLIPS'}, "
      f"significance: {'still significant' if p_topsis_loco < 0.05 else 'not significant at p<0.05'}")
```

### 2. Benefit-framing vs. cost-framing

The primary specification treats water, waste, external-input and land-use-fallow as
**costs/burdens** — lower is better, hence the `1 − x` flip (settled 2026-09-12). The
alternative tested here treats every indicator as a **benefit** instead — higher raw value
is always better, no flips at all — using the `_norm` columns directly. This is a deliberate
sensitivity check on a modeling *convention*, not new evidence: since `municipal_burden`
drives the entire causal result, and benefit-framing is specifically the flip this category
depends on, **this check is expected to change the sign of the DiD estimate by
construction**, not to discover something new about food sovereignty. Reported for
completeness, per the note's standing commitment to carry it as an appendix-level check.

```python
def build_fsoi_benefit_framed(df, category_map, suffix, year_range=None):
    """No cost-direction flips at all -- every indicator uses its _norm column directly
    (higher raw value always scores higher), unlike the primary spec's _score columns.
    Same category-then-within equal weighting; only the direction convention differs."""
    if year_range is not None:
        lo, hi = year_range
        df = df[df["Year"].between(lo, hi)]
    out = df[["Year", "Location_Name", "Treated", "Treated_Label"]].copy()
    for category, indicators in category_map.items():
        cols = [f"{ind}{suffix}_norm" for ind in indicators]
        out[category] = df[cols].mean(axis=1)
    out["FSOI"] = out[list(category_map)].mean(axis=1)
    return out


FSOI_A_benefit = build_fsoi_benefit_framed(data_official_Türkiye, CATEGORY_MAP_A, suffix="_perHousehold", year_range=None)
assert FSOI_A_benefit["FSOI"].between(0, 1, inclusive="both").all(), "FSOI_A_benefit out of [0, 1]"

did_sample_benefit = FSOI_A_benefit[FSOI_A_benefit["Treated_Label"].isin(
    ["Non-metropolitan", "New-metropolitan (2012)"])].copy()
did_sample_benefit["Treated_bin"] = (did_sample_benefit["Treated_Label"] == "New-metropolitan (2012)").astype(int)
did_sample_benefit["Post"] = (did_sample_benefit["Year"] >= 2014).astype(int)
did_sample_benefit["TreatedXPost"] = did_sample_benefit["Treated_bin"] * did_sample_benefit["Post"]

did_model_benefit = smf.ols("FSOI ~ TreatedXPost + C(Location_Name) + C(Year)", data=did_sample_benefit).fit(
    cov_type="cluster", cov_kwds={"groups": did_sample_benefit["Location_Name"]})
coef_benefit, p_benefit = did_model_benefit.params["TreatedXPost"], did_model_benefit.pvalues["TreatedXPost"]
print(f"PRIMARY DiD, benefit-framed (no cost flips): {coef_benefit:+.4f} "
      f"(SE {did_model_benefit.bse['TreatedXPost']:.4f}, p = {p_benefit:.4f})")
print(f"PRIMARY DiD, cost-framed (settled primary):  {did_coef:+.4f} (SE {did_se:.4f}, p = {did_p:.4f})")
print(f"\nSign {'flips as expected by construction' if (coef_benefit < 0) != (did_coef < 0) else 'does NOT flip -- worth a closer look, this was not expected'}"
      f" -- benefit-framing directly re-signs the municipal_burden indicators the DiD result depends on.")
print("This is not new evidence about food sovereignty: it demonstrates the result's sign is")
print("a direct, mechanical consequence of the cost/burden framing convention, which was")
print("settled deliberately (2026-09-12) and is not being reopened by this check.")
```

### 3. Waiver-years exclusion

Re-estimates the primary DiD dropping the three waiver-window years (2014, 2016, 2018) from
the sample entirely, keeping pre-treatment (2008, 2010, 2012) and only the post-waiver years
(2020, 2022, 2024) as "post." Direct re-estimation test of whether the effect depends on the
waiver-affected years specifically, complementing (not repeating) the event-study shape
check above.

```python
did_sample_no_waiver = did_sample[~did_sample["Year"].isin([2014, 2016, 2018])].copy()
print(f"Waiver-years-excluded sample: {len(did_sample_no_waiver)} obs, "
      f"years {sorted(did_sample_no_waiver['Year'].unique())}")

did_model_no_waiver = smf.ols("FSOI ~ TreatedXPost + C(Location_Name) + C(Year)", data=did_sample_no_waiver).fit(
    cov_type="cluster", cov_kwds={"groups": did_sample_no_waiver["Location_Name"]})
coef_nw, se_nw, p_nw = (did_model_no_waiver.params["TreatedXPost"], did_model_no_waiver.bse["TreatedXPost"],
                         did_model_no_waiver.pvalues["TreatedXPost"])
print(f"\nDiD, waiver years excluded (post = 2020/2022/2024 only): {coef_nw:+.4f} "
      f"(SE {se_nw:.4f}, p = {p_nw:.4f}, n = {int(did_model_no_waiver.nobs)})")
print(f"DiD, full sample (primary, for comparison):                {did_coef:+.4f} (SE {did_se:.4f}, p = {did_p:.4f})")

# LOCO in the waiver-excluded sample too, for completeness.
did_sample_loco_no_waiver = did_sample_loco[~did_sample_loco["Year"].isin([2014, 2016, 2018])].copy()
did_model_loco_nw = smf.ols("FSOI ~ TreatedXPost + C(Location_Name) + C(Year)", data=did_sample_loco_no_waiver).fit(
    cov_type="cluster", cov_kwds={"groups": did_sample_loco_no_waiver["Location_Name"]})
coef_loco_nw = did_model_loco_nw.params["TreatedXPost"]
p_loco_nw = did_model_loco_nw.pvalues["TreatedXPost"]
print(f"\nWithout municipal_burden, waiver years excluded: {coef_loco_nw:+.4f} (p = {p_loco_nw:.4f})")
print(f"Sign check: {'same direction as full-sample LOCO' if (coef_loco_nw < 0) == (coef_loco < 0) else 'differs from full-sample LOCO'}")
```

### Summary — final robustness batch

**1. TOPSIS — the causal finding gets even stronger confirmation, but a descriptive claim
does not survive.** City-level ranks agree closely with equal-weighted-sum (ρ = 0.917 — much
tighter than flat-weighting's 0.681, expected since TOPSIS reuses the identical weights and
only changes the aggregation formula). The DiD result replicates: −0.025 (p = 0.002) vs. the
primary −0.037 (p = 0.003), and **without municipal_burden it is −0.0003 (p = 0.968) — not
just insignificant but essentially exactly zero.** This is the sharpest confirmation yet,
across a third independent method, that the COMBINED remaining three categories show no
significant net effect (individually, land-use and external-input each carry their own
small, partly-offsetting effect — see the 2026-09-24 correction above).

**⚠️ But TOPSIS does not reproduce "non-metropolitan scores highest."** Group means: equal-
weighted-sum gives non-metro 0.507 > old-metro 0.484 > new-metro 0.466. **TOPSIS gives
old-metro 0.366 > non-metro 0.336 > new-metro 0.322 — old-metropolitan leads, not
non-metropolitan.** New-metropolitan stays lowest under both methods, which is what the DiD
actually needs (it compares new-metro to non-metro, not to "whoever is highest"), so **the
causal estimate is unaffected** — but the earlier descriptive claim from the LOCO section
above ("non-metropolitan cities score higher... not only when municipal burden is included")
is **method-specific, not aggregation-method-robust**, and must be stated as such rather than
as a general finding. Correct framing: *new-metropolitan cities score lowest under both
aggregation methods tested; which of the other two groups leads is sensitive to the choice
between equal-weighted-sum and TOPSIS.*

**2. Benefit-framing — flips as predicted, confirms the mechanism rather than testing it.**
+0.031 (p = 0.008) vs. −0.037 cost-framed. Exactly the expected mechanical flip, since
benefit-framing directly re-signs the municipal_burden indicators the entire result depends
on. Not new evidence — it demonstrates the result's sign is a direct consequence of the
cost/burden convention settled 2026-09-12, which this check does not reopen.

**3. Waiver-years exclusion — effect holds outside the waiver window, at reduced power.**
Dropping 2014/2016/2018 entirely (n: 585 → 390): −0.034 (p = 0.042) vs. −0.037 (p = 0.003)
full-sample — similar magnitude, weaker significance (smaller n, fewer post-treatment
points), but the effect does not depend on the waiver years specifically. Without
municipal_burden: +0.003 (p = 0.746) — same story, a fourth time.

**Where this leaves the whole DiD analysis, across four independent checks (category-
weighted, flat-weighted, TOPSIS, waiver-years-excluded):** the significant negative effect on
the composite, and its complete disappearance without municipal_burden, both replicate
robustly. The one genuinely new finding from this batch is that **the descriptive
group-ordering claim (which non-metropolitan-vs-old-metropolitan group is "highest") is not
aggregation-method-robust and should not be reported as a settled fact** — only "new-metro
lowest" and "the causal effect is a municipal_burden effect, not a food-sovereignty effect"
survive across every specification tried.

## Does the original "national decline, independent of Law 6360" finding survive?

**Orhan's question (2026-09-24), a different question from everything in the robustness
queue above.** The pre-rebuild finding was that FSOI declines across **all three groups**
over time, interpreted as a national trend unrelated to the reform — exactly the kind of
common trend the DiD design nets out via year fixed effects, so nothing checked this week
actually tested whether it still holds on the rebuilt index.

**Answered directly via group × year means, not by extracting year-FE coefficients from the
DiD models above** — those models (`did_model`, `did_model_loco`) are fit on the
non-metropolitan-vs-new-metropolitan sample only and cannot speak to old-metropolitan's
trend at all, which the original claim explicitly included. `FSOI_A` and `FSOI_A_loco_mb`
are both already built on the full panel (all three real groups), so comparing their
group × year trajectories directly answers "does the decline persist without
municipal_burden" for every group, not just the two the DiD uses.

```python
trend_full = (FSOI_A[FSOI_A["Location_Name"] != "Türkiye"]
              .groupby(["Year", "Treated_Label"], observed=True)["FSOI"].mean().unstack())
trend_loco = (FSOI_A_loco_mb[FSOI_A_loco_mb["Location_Name"] != "Türkiye"]
              .groupby(["Year", "Treated_Label"], observed=True)["FSOI"].mean().unstack())

print("Group means by year -- FULL 4-category Track A (with municipal_burden):")
print(trend_full.round(3).to_string())
print("\nGroup means by year -- 3-category, municipal_burden EXCLUDED (production + land_use + external_input):")
print(trend_loco.round(3).to_string())

print("\nFirst year (2008) vs last year (2024), each group, each version:")
for label, trend in [("WITH municipal_burden", trend_full), ("WITHOUT municipal_burden", trend_loco)]:
    print(f"\n  {label}:")
    for group in trend.columns:
        start, end = trend[group].iloc[0], trend[group].iloc[-1]
        change = end - start
        print(f"    {group:28s} 2008: {start:.3f} -> 2024: {end:.3f}  (change: {change:+.3f}, "
              f"{'decline' if change < 0 else 'increase'})")
```

```python
# One clean number per version: a common linear time trend, city fixed effects, no
# treatment terms at all -- literally "is there a national decline" as a single coefficient,
# tested with and without municipal_burden.
trend_sample_full = FSOI_A[FSOI_A["Location_Name"] != "Türkiye"].copy()
trend_sample_loco = FSOI_A_loco_mb[FSOI_A_loco_mb["Location_Name"] != "Türkiye"].copy()

trend_model_full = smf.ols("FSOI ~ Year + C(Location_Name)", data=trend_sample_full).fit(
    cov_type="cluster", cov_kwds={"groups": trend_sample_full["Location_Name"]})
trend_model_loco = smf.ols("FSOI ~ Year + C(Location_Name)", data=trend_sample_loco).fit(
    cov_type="cluster", cov_kwds={"groups": trend_sample_loco["Location_Name"]})

print("Common linear time trend (per year, all three groups pooled, city FE, cluster-robust SE):")
print(f"  WITH municipal_burden (4 categories):    {trend_model_full.params['Year']:+.5f} "
      f"(SE {trend_model_full.bse['Year']:.5f}, p = {trend_model_full.pvalues['Year']:.4f})")
print(f"  WITHOUT municipal_burden (3 categories): {trend_model_loco.params['Year']:+.5f} "
      f"(SE {trend_model_loco.bse['Year']:.5f}, p = {trend_model_loco.pvalues['Year']:.4f})")

both_negative = trend_model_full.params["Year"] < 0 and trend_model_loco.params["Year"] < 0
both_significant = trend_model_full.pvalues["Year"] < 0.05 and trend_model_loco.pvalues["Year"] < 0.05
print()
if both_negative and both_significant:
    print("DECLINE PERSISTS without municipal_burden -- consistent with a real, separate national")
    print("agricultural-self-sufficiency finding, distinct from the Law 6360 DiD story.")
elif both_negative and not trend_model_loco.pvalues["Year"] < 0.05:
    print("Direction persists (still negative) but loses significance without municipal_burden --")
    print("weaker evidence for the national-decline claim once the confounded category is removed.")
else:
    print("Decline does NOT clearly persist without municipal_burden -- the original national-decline")
    print("finding may itself have been partly carried by municipal_burden, the same way the")
    print("Law 6360 causal estimate was. Needs the same qualification as the group-ranking correction.")
```

### Reading this, once run

Two independent tests of the same question: the raw 2008-vs-2024 group trajectories above,
and a single pooled linear time trend here. If both agree the decline **persists** (negative,
significant) with municipal_burden removed, that licenses treating it as a real, separate
finding about national agricultural self-sufficiency — unconnected to Law 6360, since this
whole check deliberately excludes any treatment/DiD structure. If it **doesn't** persist, the
original claim needs the same kind of qualification the group-ranking claim just got: real in
the full composite, but substantially a municipal_burden artifact rather than a genuine
agricultural trend. Either answer is reportable — the point of this check is to know which,
before it goes in the thesis as either a supporting or an independent finding.

## Direct test of the waiver mechanism on water_drainage itself

The mechanism check earlier tested the waiver against `municipal_burden` (waste only, in
Track A) — the waiver's broader "no fees" language, not its headline provision. This runs
the same event-study design directly on **`water_drainage`**, the variable the waiver
actually targeted (drinking/usage water tariffs capped 2014–2019). `panel_C` (the merged
main+extended panel) carries `water_drainage` through **2022** — one biennial point further
than Track A's water-free categories, and enough to see one point past the waiver's end
independent of Track A's own year range.

```python
water_sample = panel_C[
    panel_C["Treated_Label"].isin(["Non-metropolitan", "New-metropolitan (2012)"])
    & panel_C["Year"].between(2008, 2022)
].dropna(subset=["water_drainage_perHousehold_score"]).copy()
water_sample["Treated_bin"] = (water_sample["Treated_Label"] == "New-metropolitan (2012)").astype(int)

n_water_cities = water_sample["Location_Name"].nunique()
print(f"Water sample: {len(water_sample)} obs, {n_water_cities} cities, "
      f"years {sorted(water_sample['Year'].unique())}")

water_event_years = sorted(water_sample["Year"].unique())
water_ref_year = 2012
for yr in water_event_years:
    if yr == water_ref_year:
        continue
    water_sample[f"WxY_{yr}"] = ((water_sample["Year"] == yr) & (water_sample["Treated_bin"] == 1)).astype(int)

water_terms = [f"WxY_{yr}" for yr in water_event_years if yr != water_ref_year]
water_formula = "water_drainage_perHousehold_score ~ " + " + ".join(water_terms) + " + C(Location_Name) + C(Year)"
water_event_model = smf.ols(water_formula, data=water_sample).fit(
    cov_type="cluster", cov_kwds={"groups": water_sample["Location_Name"]})

water_rows = []
for yr in water_event_years:
    if yr == water_ref_year:
        water_rows.append({"Year": yr, "coef": 0.0, "se": 0.0, "p": np.nan, "period": "ref (2012)"})
        continue
    term = f"WxY_{yr}"
    water_rows.append({
        "Year": yr, "coef": water_event_model.params[term], "se": water_event_model.bse[term],
        "p": water_event_model.pvalues[term], "period": "pre" if yr < 2014 else "post",
    })
water_event_study = pd.DataFrame(water_rows)
print("\nTreatment effect on water_drainage ALONE, relative to 2012, by year:")
print(water_event_study.round(4).to_string(index=False))
```

```python
# Primary two-period DiD on water_drainage alone, for one clean number alongside the shape.
water_sample["Post"] = (water_sample["Year"] >= 2014).astype(int)
water_sample["TreatedXPost"] = water_sample["Treated_bin"] * water_sample["Post"]
water_did_model = smf.ols("water_drainage_perHousehold_score ~ TreatedXPost + C(Location_Name) + C(Year)",
                           data=water_sample).fit(cov_type="cluster", cov_kwds={"groups": water_sample["Location_Name"]})
water_coef, water_se, water_p = (water_did_model.params["TreatedXPost"], water_did_model.bse["TreatedXPost"],
                                  water_did_model.pvalues["TreatedXPost"])
print(f"Primary DiD, water_drainage alone: {water_coef:+.4f} (SE {water_se:.4f}, p = {water_p:.4f}, n = {int(water_did_model.nobs)})")

water_pre = water_event_study[water_event_study["period"] == "pre"]["coef"]
water_2014_2018 = water_event_study[water_event_study["Year"].isin([2014, 2016, 2018])]["coef"]
water_2020_2022 = water_event_study[water_event_study["Year"].isin([2020, 2022])]["coef"]
print(f"\nMean effect, 2014-2018 (waiver active): {water_2014_2018.mean():+.4f}")
print(f"Mean effect, 2020-2022 (waiver ended):   {water_2020_2022.mean():+.4f}")
print("\nCompare to municipal_burden (waste) shape: 2014-2018 mean -0.2045, 2020-2024 mean -0.1484 (27% fade).")
if abs(water_2020_2022.mean()) < abs(water_2014_2018.mean()):
    print("Water ALSO fades after the waiver ends -- same qualitative shape as waste, now confirmed on")
    print("the variable the waiver actually targeted. Strengthens the price-driven-waiver reading.")
else:
    print("Water does NOT show the same fade as waste -- the two municipal indicators diverge, which")
    print("weakens treating this as one unified 'municipal_burden confound' story and suggests the")
    print("waste finding and any water finding may need separate explanations.")
```

## Full leave-one-category-out — every Track A category, not just municipal_burden

Only `municipal_burden` has been tested this way so far. Completing the set for Track A's
other three categories (`production`, `land_use`, `external_input`) checks whether the
causal estimate depends on any of them the way it depends on municipal_burden, or whether
municipal_burden really is the unique driver.

```python
loco_results = []

# municipal_burden already computed above (coef_loco, p_loco) -- included for a complete table.
loco_results.append({"removed_category": "municipal_burden", "coef": coef_loco, "p": p_loco})

for cat_to_remove in ["production", "land_use", "external_input"]:
    cmap = {k: v for k, v in CATEGORY_MAP_A.items() if k != cat_to_remove}
    fsoi_loco_cat = build_fsoi(data_official_Türkiye, cmap, suffix="_perHousehold", year_range=None)
    sample = fsoi_loco_cat[fsoi_loco_cat["Treated_Label"].isin(
        ["Non-metropolitan", "New-metropolitan (2012)"])].copy()
    sample["Treated_bin"] = (sample["Treated_Label"] == "New-metropolitan (2012)").astype(int)
    sample["Post"] = (sample["Year"] >= 2014).astype(int)
    sample["TreatedXPost"] = sample["Treated_bin"] * sample["Post"]
    m = smf.ols("FSOI ~ TreatedXPost + C(Location_Name) + C(Year)", data=sample).fit(
        cov_type="cluster", cov_kwds={"groups": sample["Location_Name"]})
    loco_results.append({"removed_category": cat_to_remove,
                          "coef": m.params["TreatedXPost"], "p": m.pvalues["TreatedXPost"]})

loco_table = pd.DataFrame(loco_results)
loco_table["significant"] = loco_table["p"] < 0.05
loco_table["same_sign_as_primary"] = np.sign(loco_table["coef"]) == np.sign(did_coef)
print(f"Primary DiD (all 4 categories): {did_coef:+.4f} (p = {did_p:.4f})")
print("\nDiD estimate with EACH category removed in turn:")
print(loco_table.round(4).to_string(index=False))
print("\nReportable: any row besides municipal_burden that changes sign or loses significance")
print("would mean that category ALSO drives part of the result -- check before concluding")
print("municipal_burden is the sole driver.")
```

## perArea robustness — group means, Spearman, and a perArea DiD

`FSOI_A_perArea` was built during the original aggregation pass but never taken further.
perArea is known (Diagnostic 1/6) to be ~99% population density **across** cities — a poor
descriptive/ranking track — but **clean within a city over time**, since a city's area is a
fixed denominator. That makes it worth a genuine DiD attempt even though it's the wrong track
for levels and rankings: city fixed effects absorb exactly the cross-sectional density
confound that makes perArea uninterpretable as a ranking.

```python
# Group means and Spearman: perHousehold vs perArea, Track A, pooled all years.
groups_ph = FSOI_A[FSOI_A["Location_Name"] != "Türkiye"].groupby("Treated_Label", observed=True)["FSOI"].mean().round(3)
groups_pa = FSOI_A_perArea[FSOI_A_perArea["Location_Name"] != "Türkiye"].groupby("Treated_Label", observed=True)["FSOI"].mean().round(3)
print("Group means, perHousehold (primary) vs perArea, Track A, pooled all years:")
print(pd.concat([groups_ph.rename("perHousehold"), groups_pa.rename("perArea")], axis=1))

rank_corr_perarea = spearmanr(
    FSOI_A[FSOI_A["Location_Name"] != "Türkiye"].groupby("Location_Name")["FSOI"].mean(),
    FSOI_A_perArea[FSOI_A_perArea["Location_Name"] != "Türkiye"].groupby("Location_Name")["FSOI"].mean(),
).statistic
print(f"\nCity-level rank correlation, perHousehold vs perArea (Track A, all years pooled): rho = {rank_corr_perarea:.3f}")
print("(For reference, the other cross-specification rank correlations found this session:")
print(" flat-weighting vs category-weighted: rho=0.681 | TOPSIS vs category-weighted: rho=0.917)")
```

### Clarifying the full-LOCO table: mechanical weight redistribution, or real category-level effects?

The table above shows removing `production`, `land_use`, or `external_input` *individually*
makes the DiD estimate **more** negative and more significant than the primary (all four
categories). Read naively that looks like "removing agriculture categories worsens the
agriculture effect," which would be a contradiction. The more likely explanation is
mechanical: in a 4-category equal-weighted mean, removing *any* category other than
`municipal_burden` increases municipal_burden's relative share of what's left (from 1/4 to
1/3), pulling the composite further toward whatever municipal_burden's own effect is. Tested
directly: each category run **alone**, as its own single-category "composite," isolating
each one's own DiD estimate rather than its effect on a blend.

```python
isolation_rows = []
for category in CATEGORY_MAP_A:
    solo_map = {category: CATEGORY_MAP_A[category]}
    fsoi_solo = build_fsoi(data_official_Türkiye, solo_map, suffix="_perHousehold", year_range=None)
    sample = fsoi_solo[fsoi_solo["Treated_Label"].isin(
        ["Non-metropolitan", "New-metropolitan (2012)"])].copy()
    sample["Treated_bin"] = (sample["Treated_Label"] == "New-metropolitan (2012)").astype(int)
    sample["Post"] = (sample["Year"] >= 2014).astype(int)
    sample["TreatedXPost"] = sample["Treated_bin"] * sample["Post"]
    m = smf.ols("FSOI ~ TreatedXPost + C(Location_Name) + C(Year)", data=sample).fit(
        cov_type="cluster", cov_kwds={"groups": sample["Location_Name"]})
    isolation_rows.append({"category_alone": category, "coef": m.params["TreatedXPost"],
                            "p": m.pvalues["TreatedXPost"]})

isolation_table = pd.DataFrame(isolation_rows)
isolation_table["significant"] = isolation_table["p"] < 0.05
print("Each category run ALONE as its own single-category composite (isolates each category's own effect):")
print(isolation_table.round(4).to_string(index=False))
print("\nIf production/land_use/external_input are each individually null or positive, the")
print("full-LOCO table's 'more negative when removed' pattern is mechanical weight")
print("redistribution toward municipal_burden, not evidence those categories carry their own")
print("negative effect.")
```

## Denominator decomposition — what actually separates perHousehold from perArea

The perArea/perHousehold tension (Correction 2) can be taken apart exactly, without any
normalisation in the way, by running the DiD on **logs of the raw quantities**. In logs every
denominator is additive, and the DiD is linear, so the coefficients add up:

- `log(waste / household) = log(waste) − log(household count)`
- `log(waste / capita)    = log(waste) − log(population)`
- `log(household count)   = log(population) − log(household size)`
- `log(waste / area) ≡ log(waste)` once city fixed effects are in — area never changes within a city

So one table shows *which* piece carries the effect: the numerator (waste itself), or the
denominator. **This also corrects the previous check:** the household-count event study just
above was run in raw levels, and treated cities are larger — equal percentage growth in both
groups would still produce a growing absolute gap. The log version below is the right test.

```python
decomp = diagnostic_raw[diagnostic_raw["Treated_Label"].isin(
    ["Non-metropolitan", "New-metropolitan (2012)"])].copy()
decomp["Treated_bin"] = (decomp["Treated_Label"] == "New-metropolitan (2012)").astype(int)
decomp["Post"] = (decomp["Year"] >= 2014).astype(int)
decomp["TreatedXPost"] = decomp["Treated_bin"] * decomp["Post"]
assert (decomp[["Waste_Collected_1000Ton", "Population_Total", "Mean_Household_Count",
                "Mean_Household_Size"]] > 0).all().all(), "log needs strictly positive values"

decomp["log_waste"] = np.log(decomp["Waste_Collected_1000Ton"])
decomp["log_pop"] = np.log(decomp["Population_Total"])
decomp["log_hh_count"] = np.log(decomp["Mean_Household_Count"])
decomp["log_hh_size"] = np.log(decomp["Mean_Household_Size"])
decomp["log_waste_per_hh"] = decomp["log_waste"] - decomp["log_hh_count"]
decomp["log_waste_per_cap"] = decomp["log_waste"] - decomp["log_pop"]

decomp_rows = []
for outcome, meaning in [
    ("log_waste", "waste total  (== perArea under city FE)"),
    ("log_pop", "population"),
    ("log_hh_count", "household count"),
    ("log_hh_size", "household size"),
    ("log_waste_per_hh", "waste per household"),
    ("log_waste_per_cap", "waste per capita"),
]:
    m = smf.ols(f"{outcome} ~ TreatedXPost + C(Location_Name) + C(Year)", data=decomp).fit(
        cov_type="cluster", cov_kwds={"groups": decomp["Location_Name"]})
    decomp_rows.append({"outcome": outcome, "meaning": meaning,
                        "coef": m.params["TreatedXPost"], "se": m.bse["TreatedXPost"],
                        "p": m.pvalues["TreatedXPost"],
                        "approx_%_change": 100 * (np.exp(m.params["TreatedXPost"]) - 1)})
decomp_table = pd.DataFrame(decomp_rows).set_index("outcome")
print("DiD on logs of raw quantities (non-metro vs new-metro, two-way FE, cluster-robust SE):")
print(decomp_table.round(4).to_string())

# The identities must hold exactly (linear model, identical sample) -- a check that the
# decomposition is arithmetic, not interpretation.
c = decomp_table["coef"]
assert abs(c["log_waste_per_hh"] - (c["log_waste"] - c["log_hh_count"])) < 1e-8
assert abs(c["log_waste_per_cap"] - (c["log_waste"] - c["log_pop"])) < 1e-8
assert abs(c["log_hh_count"] - (c["log_pop"] - c["log_hh_size"])) < 1e-8
print("\nAdditivity identities verified: per-household = waste - household count; "
      "per-capita = waste - population; household count = population - household size.")
```

```python
# Household count and population event studies IN LOGS -- the correct scale for "is there
# a jump or a differential trend", unlike the raw-level version above.
log_event_rows = []
for outcome in ["log_hh_count", "log_pop", "log_hh_size", "log_waste"]:
    for yr in hh_event_years:
        if yr != hh_ref_year:
            decomp[f"LxY_{yr}"] = ((decomp["Year"] == yr) & (decomp["Treated_bin"] == 1)).astype(int)
    terms = [f"LxY_{yr}" for yr in hh_event_years if yr != hh_ref_year]
    m = smf.ols(f"{outcome} ~ " + " + ".join(terms) + " + C(Location_Name) + C(Year)", data=decomp).fit(
        cov_type="cluster", cov_kwds={"groups": decomp["Location_Name"]})
    for yr in hh_event_years:
        if yr == hh_ref_year:
            log_event_rows.append({"outcome": outcome, "Year": yr, "coef": 0.0, "p": np.nan})
        else:
            log_event_rows.append({"outcome": outcome, "Year": yr,
                                   "coef": m.params[f"LxY_{yr}"], "p": m.pvalues[f"LxY_{yr}"]})
log_event = pd.DataFrame(log_event_rows)
print("Event studies in LOGS, relative to 2012 (coef x 100 ~ percent difference):")
print(log_event.pivot(index="Year", columns="outcome", values="coef").round(4).to_string())
print("\np-values:")
print(log_event.pivot(index="Year", columns="outcome", values="p").round(3).to_string())
```

## A per-capita track

`perHousehold = total / household count = total × household size / population`, so
`per-capita = perHousehold / household size` exactly. Built for every Track A indicator
with one exception: `landuse_core` is already a unitless mean of two min-max-scaled columns
(harvested and sown), so dividing it by household size would not be a per-capita quantity.
It is replaced here by harvested land per capita from `diagnostic_raw` — harvested and sown
correlate r = 0.99, which is why they were collapsed in the first place. Normalised with the
identical recipe (log1p → winsorise 1/99 on cities → pooled min-max), which is first checked
by reproducing an existing perHousehold `_norm` column exactly.

```python
def normalise_one(series, is_city):
    """Same recipe as normalise_indicators(), for a single column."""
    s = np.log1p(series) if series[is_city].skew() > LOG1P_SKEW_THRESHOLD else series.copy()
    lo, hi = s[is_city].quantile(WINSOR_LIMITS[0]), s[is_city].quantile(WINSOR_LIMITS[1])
    return (s.clip(lo, hi) - lo) / (hi - lo)

is_city_main = data_official_Türkiye["Location_Name"] != "Türkiye"

# Verify the re-implementation reproduces an existing normalised column before trusting it.
_check = normalise_one(data_official_Türkiye["wasteCollected_1000ton_perHousehold"], is_city_main)
assert np.allclose(_check, data_official_Türkiye["wasteCollected_1000ton_perHousehold_norm"], equal_nan=True), \
    "normalise_one does not reproduce the pipeline's normalisation"

pc = data_official_Türkiye[["Year", "Location_Name"]].merge(
    diagnostic_raw[["Year", "Location_Name", "Mean_Household_Size", "Population_Total", "Landuse_Harvest_km2"]],
    on=["Year", "Location_Name"], how="left", validate="one_to_one")
assert pc["Mean_Household_Size"].notna().all(), "household size missing for some rows"

for ind in MAIN_INDICATORS:
    if ind == "landuse_core":
        raw_pc = (pc["Landuse_Harvest_km2"] / pc["Population_Total"]).values
    else:
        raw_pc = (data_official_Türkiye[f"{ind}_perHousehold"].values / pc["Mean_Household_Size"].values)
    data_official_Türkiye[f"{ind}_perCapita"] = raw_pc
    normed = normalise_one(data_official_Türkiye[f"{ind}_perCapita"], is_city_main)
    data_official_Türkiye[f"{ind}_perCapita_norm"] = normed
    data_official_Türkiye[f"{ind}_perCapita_score"] = (1 - normed) if ind in COST_INDICATORS else normed

# Cross-check the algebra on waste: perHousehold / size must equal total / population.
_direct = diagnostic_raw.set_index(["Year", "Location_Name"])
_direct = (_direct["Waste_Collected_1000Ton"] / _direct["Population_Total"])
_via = data_official_Türkiye.set_index(["Year", "Location_Name"])["wasteCollected_1000ton_perCapita"]
assert np.allclose(_via.loc[_direct.index], _direct, rtol=1e-6), "per-capita algebra check failed"
print("normalise_one reproduces the pipeline exactly; per-capita = perHousehold / size verified against total / population.")

FSOI_A_perCapita = build_fsoi(data_official_Türkiye, CATEGORY_MAP_A, suffix="_perCapita", year_range=None)
assert FSOI_A_perCapita["FSOI"].between(0, 1).all()
print(f"FSOI_A_perCapita built: {FSOI_A_perCapita.shape[0]} rows")
```

## Exhaustive isolation and LOCO across all three denominators

For each denominator: the full four-category DiD, each category **alone**, and each
category **removed**. One table, so the three tracks can be read against each other row by
row.

```python
def did_estimate(fsoi_df):
    s = fsoi_df[fsoi_df["Treated_Label"].isin(["Non-metropolitan", "New-metropolitan (2012)"])].copy()
    s["TreatedXPost"] = ((s["Treated_Label"] == "New-metropolitan (2012)") & (s["Year"] >= 2014)).astype(int)
    m = smf.ols("FSOI ~ TreatedXPost + C(Location_Name) + C(Year)", data=s).fit(
        cov_type="cluster", cov_kwds={"groups": s["Location_Name"]})
    return m.params["TreatedXPost"], m.pvalues["TreatedXPost"]

exhaustive_rows = []
for suffix in ["_perHousehold", "_perArea", "_perCapita"]:
    specs = [("all four", CATEGORY_MAP_A)]
    specs += [(f"{c} alone", {c: CATEGORY_MAP_A[c]}) for c in CATEGORY_MAP_A]
    specs += [(f"without {c}", {k: v for k, v in CATEGORY_MAP_A.items() if k != c}) for c in CATEGORY_MAP_A]
    for label, cmap in specs:
        coef, p = did_estimate(build_fsoi(data_official_Türkiye, cmap, suffix=suffix, year_range=None))
        exhaustive_rows.append({"spec": label, "denominator": suffix.strip("_"),
                                "coef": coef, "p": p})

exhaustive = pd.DataFrame(exhaustive_rows)
exhaustive["cell"] = exhaustive.apply(
    lambda r: f"{r.coef:+.4f}{'*' if r.p < 0.05 else ' '} (p={r.p:.3f})", axis=1)
print("DiD coefficient by specification and denominator (* = p < 0.05):")
print(exhaustive.pivot(index="spec", columns="denominator", values="cell")
      .reindex([s for s in dict.fromkeys(exhaustive["spec"])])
      [["perHousehold", "perCapita", "perArea"]].to_string())
```

### Three checks the reading above depends on

1. **Scale compression** — if perArea's normalised range is dominated by the cross-city
   density spread, within-city movement will be a small fraction of it, shrinking every
   perArea DiD coefficient regardless of what actually happened.
2. **The fade, in logs** — the normalised-score event studies showed a 27–34% fade after
   2019 that was read as favouring the tariff-waiver mechanism. Winsorising and min-max are
   non-linear at the edges; the log scale is the honest test.
3. **Coverage** — Law 6360 extended municipal services province-wide. If the waste effect is
   that extension, the log jump in waste collected should match the log jump in the implied
   municipal coverage share (computed from TÜİK's own per-person and total water figures,
   the same inversion used in Diagnostic 2).

```python
# 1. Scale compression: within-city SD (city-demeaned) as a share of the full score range.
sc_rows = []
for suffix in ["_perHousehold", "_perCapita", "_perArea"]:
    col = f"wasteCollected_1000ton{suffix}_score"
    s = data_official_Türkiye.loc[is_city_main, ["Location_Name", col]].dropna()
    within = s[col] - s.groupby("Location_Name")[col].transform("mean")
    between = s.groupby("Location_Name")[col].mean()
    sc_rows.append({"denominator": suffix.strip("_"), "within_city_sd": within.std(),
                    "between_city_sd": between.std(),
                    "within_share_of_total_var": within.var() / s[col].var()})
scale_table = pd.DataFrame(sc_rows).set_index("denominator")
print("Waste score: how much of the variation is within-city (what a DiD uses) vs between-city:")
print(scale_table.round(4).to_string())
ratio = scale_table.loc["perHousehold", "within_city_sd"] / scale_table.loc["perArea", "within_city_sd"]
print(f"\nWithin-city SD is {ratio:.1f}x larger under perHousehold than perArea.")
mb = exhaustive[exhaustive["spec"] == "municipal_burden alone"].set_index("denominator")["coef"]
print(f"municipal_burden-alone DiD ratio (perHousehold / perArea): {mb['perHousehold'] / mb['perArea']:.1f}x")
```

```python
# 2 and 3. Water and implied municipal coverage share, in logs, event-study and DiD.
wc = diagnostic_raw[diagnostic_raw["Treated_Label"].isin(
    ["Non-metropolitan", "New-metropolitan (2012)"])].copy()
wc["Treated_bin"] = (wc["Treated_Label"] == "New-metropolitan (2012)").astype(int)
wc["TreatedXPost"] = wc["Treated_bin"] * (wc["Year"] >= 2014).astype(int)
# implied municipal population = annual litres / (litres per person per day * 365)
wc["implied_muni_pop"] = (wc["Water_Drainage_1000m3PerYear"] * 1e6) / (wc["Water_Drainage_LitrePerPersonPerDay"] * 365)
wc["coverage_share"] = wc["implied_muni_pop"] / wc["Population_Total"]
wc["log_water"] = np.log(wc["Water_Drainage_1000m3PerYear"])
wc["log_coverage"] = np.log(wc["coverage_share"])
wc["log_waste"] = np.log(wc["Waste_Collected_1000Ton"])

rows, ev = [], []
for outcome in ["log_waste", "log_water", "log_coverage"]:
    d = wc.dropna(subset=[outcome]).copy()
    d = d[np.isfinite(d[outcome])]
    m = smf.ols(f"{outcome} ~ TreatedXPost + C(Location_Name) + C(Year)", data=d).fit(
        cov_type="cluster", cov_kwds={"groups": d["Location_Name"]})
    rows.append({"outcome": outcome, "coef": m.params["TreatedXPost"], "p": m.pvalues["TreatedXPost"],
                 "n": int(m.nobs), "years": f"{d.Year.min()}-{d.Year.max()}"})
    yrs = sorted(d["Year"].unique())
    for yr in yrs:
        if yr != 2012:
            d[f"E_{yr}"] = ((d["Year"] == yr) & (d["Treated_bin"] == 1)).astype(int)
    me = smf.ols(f"{outcome} ~ " + " + ".join(f"E_{y}" for y in yrs if y != 2012)
                 + " + C(Location_Name) + C(Year)", data=d).fit(
        cov_type="cluster", cov_kwds={"groups": d["Location_Name"]})
    for yr in yrs:
        ev.append({"outcome": outcome, "Year": yr,
                   "coef": 0.0 if yr == 2012 else me.params[f"E_{yr}"]})

coverage_did = pd.DataFrame(rows).set_index("outcome")
print("Log DiD, non-metro vs new-metro:")
print(coverage_did.round(4).to_string())
coverage_event = pd.DataFrame(ev).pivot(index="Year", columns="outcome", values="coef")
print("\nLog event studies, relative to 2012:")
print(coverage_event.round(3).to_string())

for outcome in ["log_waste", "log_water"]:
    e = coverage_event[outcome].dropna()
    early = e.loc[[y for y in [2014, 2016, 2018] if y in e.index]].mean()
    late = e.loc[[y for y in [2020, 2022, 2024] if y in e.index]].mean()
    print(f"{outcome}: 2014-18 mean {early:+.3f}, 2020+ mean {late:+.3f}, fade {100*(1-late/early):.0f}%")

print(f"\nWaste log DiD {coverage_did.loc['log_waste','coef']:+.3f} vs coverage-share log DiD "
      f"{coverage_did.loc['log_coverage','coef']:+.3f} (ratio {coverage_did.loc['log_waste','coef']/coverage_did.loc['log_coverage','coef']:.2f})")
```

### Summary — denominator decomposition (2026-09-24)

**1. Correction 2 is resolved: the perArea null is a scaling artefact, not evidence against
the effect.** On the raw log scale, where normalisation cannot interfere, waste collected
rose **+0.260 log points (~30%)** in new-metropolitan cities relative to non-metropolitan
ones (p < 0.0001). Waste per household rose ~27%, waste per capita ~23% — all significant.
Under city fixed effects, waste per area *is* raw waste, so perArea's raw signal is the
largest of all. It vanishes only after normalisation: perArea's pooled min-max range is
dominated by the 288× cross-city density spread, leaving just **0.9%** of the waste score's
variation within cities (vs 34% under perHousehold). The 10.7× smaller within-city spread
matches the 10.9× smaller perArea coefficient almost exactly. Even so, under perArea
`municipal_burden` alone is still significant in the same direction (−0.016, p < 0.001).
**Practical consequence: normalised perArea is not a usable DiD track.** Diagnostic 6 is
right that raw perArea is clean within a city; the pooled normalisation is what removes that
advantage.

**2. The denominator is not the story.** Household count moved +2% (p = 0.37), household
size +3% (p = 0.21). The effect is in the numerator. The per-capita track replicates the
perHousehold result almost exactly (−0.0369 vs −0.0373), as expected when household size has
no differential trend.

**3. The waste effect is municipal coverage expansion.** The log DiD on waste collected
(+0.260) and on the implied municipal coverage share (+0.259) are equal (ratio 1.00), and
their event studies move together year by year: flat before 2012, a jump of about +0.25 at
2014, held through 2022. The two series are independent — coverage comes from TÜİK's water
figures, waste is its own series. **Waste collected per covered resident did not change.**
Law 6360 extended municipal services to the whole province, and municipalities recorded
waste from roughly 30% more people. This is the same mechanism that removed the per-person
water series on 2026-09-20 (confound (a)), now showing up in waste's numerator.

**4. What this means for the headline.** The single significant causal result in this
pipeline measures a change in **administrative reach** — who the municipality serves and
records — not a change in household burden, and not food sovereignty. Production shows no
effect. Land use (−0.023) and external input (+0.048) show small, opposite, real effects,
replicated under per-capita. **Defensible sentence:** *Law 6360's clearest statistical
footprint is the extension of municipal service coverage; the agricultural indicators show
no large effect.*

**5. The waiver mechanism, re-read.** The 27–34% fade that favoured the waiver came from the
normalised scale. In logs, waste fades 3% (effectively a permanent level shift — what
coverage expansion predicts) and water fades 23%. Water's residual over coverage is not
cleanly testable, because the coverage share is derived from the water total. The waiver
remains possible for water. It is no longer supported by waste.
